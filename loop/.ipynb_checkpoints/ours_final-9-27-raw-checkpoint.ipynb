生成ECG的模型

```python
import os
import json
import math
import random
import time
from pathlib import Path

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader


# =========================================================
# 0. 全局配置
# =========================================================
CFG = {
    "seed": 0,
    "train_folder": "/home/ubuntu-user/Desktop/code-9-8/spilt-9-27-raw/train",
    "eval_folder": "/home/ubuntu-user/Desktop/code-9-8/spilt-9-27-raw/val",

    "artifact_dir": "/home/ubuntu-user/Desktop/code-9-8/our-9-27-raw",

    # 数据形状
    "signal_length": 640,
    "rcg_channels": 1,
    "ecg_channels": 1,

    # 训练参数
    "batch_size": 128,
    "num_epochs": 100,
    "learning_rate": 1e-4,
    "num_workers": 0,

    # 扩散参数
    "T": 100,
    "beta_start": 1e-4,
    "beta_end": 0.02,

    # 模型参数
    "in_channels": 2,          # cond(rcg) + noisy_ecg
    "base_channels": 64,
    "time_emb_dim": 128,

    # 评估参数
    "eval_subset_size": 8,
    "save_every": 10
}


# =========================================================
# 1. 随机种子 & 工具函数
# =========================================================
def set_seed(seed=0):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


set_seed(CFG["seed"])


def format_seconds(sec):
    h = int(sec // 3600)
    m = int((sec % 3600) // 60)
    s = sec % 60
    return f"{h:02d}:{m:02d}:{s:05.2f}"


# =========================================================
# 2. 目录准备
# =========================================================
ART_DIR = Path(CFG["artifact_dir"])
ART_DIR.mkdir(parents=True, exist_ok=True)


# =========================================================
# 3. 数据加载
# =========================================================
def load_data_from_folder(folder):
    """
    从指定文件夹中加载成对的 ecg 和 rcg 数据文件（每个文件作为一个独立序列）。
    文件命名格式要求为 "{数字}.csv"，每列需包含 "ECG" 和 "RCG_ch1" 标题。
    返回：ecg_list（每个元素是一个一维数组），rcg_list，prefixes
    """
    prefixes = []
    for file in os.listdir(folder):
        if file.endswith(".csv"):
            try:
                prefix = int(file.split(".")[0])
                prefixes.append(prefix)
            except ValueError:
                print(f"[WARN] 文件 {file} 命名异常（无法转为数字），跳过。")

    prefixes = sorted(list(set(prefixes)))

    ecg_list = []
    rcg_list = []

    for prefix in prefixes:
        csv_path = os.path.join(folder, f"{prefix}.csv")

        if not os.path.exists(csv_path):
            print(f"[WARN] 文件 {csv_path} 不存在，跳过。")
            continue

        try:
            data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
        except Exception as e:
            print(f"[WARN] 读取文件 {csv_path} 失败：{e}，跳过。")
            continue

        if 'ECG' not in data.dtype.names or 'RCG' not in data.dtype.names:
            print(f"[WARN] 文件 {csv_path} 中缺少 'ECG' 或 'RCG_ch1' 列，跳过。")
            continue

        ecg_data = data['ECG']
        rcg_data = data['RCG']

        if ecg_data.shape[0] != rcg_data.shape[0]:
            raise ValueError(f"[ERROR] {csv_path} 中 ECG 与 RCG 样本数不一致")

        ecg_list.append(ecg_data)
        rcg_list.append(rcg_data)

    if len(ecg_list) == 0:
        raise RuntimeError(f"[ERROR] 在文件夹 {folder} 中没有加载到有效数据")

    return ecg_list, rcg_list, prefixes


def sliding_window_cut(data, window_len, step=30):
    """
    将一维数据 data 切成长度为 window_len 的片段，步长为 step。
    返回所有片段的列表。
    """
    if len(data) < window_len:
        return []
    indices = range(0, len(data) - window_len + 1, step)
    return [data[i:i + window_len] for i in indices]


def ensure_channel_dim(x):
    """
    若输入为 (N, L)，转成 (N, 1, L)
    """
    if x.ndim == 2:
        x = x[:, None, :]
    return x


class RCGEcgDataset(Dataset):
    def __init__(self, rcg_tensor, ecg_tensor):
        self.rcg = rcg_tensor
        self.ecg = ecg_tensor

    def __len__(self):
        return len(self.rcg)

    def __getitem__(self, idx):
        return self.rcg[idx], self.ecg[idx]


# =========================================================
# 4. 时间步嵌入
# =========================================================
def sinusoidal_time_embedding(timesteps, embed_dim):
    """
    timesteps: [B]
    return: [B, embed_dim]
    """
    device = timesteps.device
    half_dim = embed_dim // 2

    exp_range = torch.arange(half_dim, device=device, dtype=torch.float32)
    exp_range = -math.log(10000.0) * exp_range / float(half_dim)

    sinusoid_inp = timesteps.float()[:, None] * torch.exp(exp_range)[None, :]
    emb = torch.cat([torch.sin(sinusoid_inp), torch.cos(sinusoid_inp)], dim=1)
    return emb


# =========================================================
# 5. 模型模块
# =========================================================
class SqueezeExcitation(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        reduced_channels = max(1, channels // reduction)
        self.avg_pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, reduced_channels),
            nn.SiLU(),
            nn.Linear(reduced_channels, channels),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _ = x.size()
        y = self.avg_pool(x).view(b, c)
        y = self.fc(y).view(b, c, 1)
        return x * y


class SelfAttention1D(nn.Module):
    def __init__(self, channels):
        super().__init__()

        if channels < 8:
            num_heads = 1
        elif channels % 8 == 0:
            num_heads = 8
        else:
            num_heads = max(1, channels // 2)

        if channels >= 8 and channels % 8 == 0:
            groups = 8
        elif channels >= 4 and channels % 4 == 0:
            groups = 4
        elif channels >= 2 and channels % 2 == 0:
            groups = 2
        else:
            groups = 1

        self.norm = nn.GroupNorm(num_groups=groups, num_channels=channels)
        self.attn = nn.MultiheadAttention(
            embed_dim=channels,
            num_heads=num_heads,
            batch_first=True
        )

    def forward(self, x):
        h = self.norm(x)
        h = h.permute(0, 2, 1)
        attn_out, _ = self.attn(h, h, h)
        attn_out = attn_out.permute(0, 2, 1)
        return x + attn_out


class ResidualBlock(nn.Module):
    def __init__(self, in_channels, out_channels, time_emb_dim=None):
        super().__init__()

        def make_norm(channels):
            if channels >= 8 and channels % 8 == 0:
                groups = 8
            elif channels >= 4 and channels % 4 == 0:
                groups = 4
            elif channels >= 2 and channels % 2 == 0:
                groups = 2
            else:
                groups = 1
            return nn.GroupNorm(num_groups=groups, num_channels=channels)

        self.norm1 = make_norm(in_channels)
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size=3, padding=1)

        self.time_fc = nn.Linear(time_emb_dim, out_channels) if time_emb_dim is not None else None

        self.norm2 = make_norm(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size=3, padding=1)

        self.se = SqueezeExcitation(out_channels)
        self.shortcut = nn.Conv1d(in_channels, out_channels, kernel_size=1) if in_channels != out_channels else None
        self.act = nn.SiLU()

    def forward(self, x, time_emb=None):
        h = self.norm1(x)
        h = self.act(h)
        h = self.conv1(h)

        if self.time_fc is not None and time_emb is not None:
            time_bias = self.time_fc(time_emb)
            h = h + time_bias[..., None]

        h = self.norm2(h)
        h = self.act(h)
        h = self.conv2(h)

        h = self.se(h)

        x_short = self.shortcut(x) if self.shortcut is not None else x
        return x_short + h


class ConditionalDiffusionModel(nn.Module):
    def __init__(self, in_channels=2, base_channels=64, time_emb_dim=128, num_timesteps=100):
        super().__init__()
        self.in_channels = in_channels
        self.base_channels = base_channels
        self.time_emb_dim = time_emb_dim
        self.num_timesteps = num_timesteps

        time_mlp_dim = base_channels * 8
        self.time_mlp = nn.Sequential(
            nn.Linear(time_emb_dim, time_mlp_dim),
            nn.SiLU(),
            nn.Linear(time_mlp_dim, time_mlp_dim)
        )

        self.init_conv = nn.Conv1d(in_channels, base_channels, kernel_size=3, padding=1)

        self.down_blocks = nn.ModuleList()
        self.downsamples = nn.ModuleList()

        down_channels = [
            base_channels,
            base_channels * 2,
            base_channels * 4,
            base_channels * 8,
            base_channels * 8,
            base_channels * 8
        ]

        in_ch = base_channels
        for i, out_ch in enumerate(down_channels):
            self.down_blocks.append(
                ResidualBlock(in_ch, out_ch, time_emb_dim=time_mlp_dim)
            )
            if i != len(down_channels) - 1:
                self.downsamples.append(
                    nn.Conv1d(out_ch, out_ch, kernel_size=4, stride=2, padding=1)
                )
            in_ch = out_ch

        self.attn = SelfAttention1D(down_channels[-1])

        self.up_blocks = nn.ModuleList()
        self.upsamples = nn.ModuleList()

        up_channels = [
            base_channels * 8,
            base_channels * 4,
            base_channels * 2,
            base_channels,
            base_channels
        ]

        in_ch = down_channels[-1]
        for i, out_ch in enumerate(up_channels):
            skip_ch = down_channels[-2 - i]
            res_in = in_ch + skip_ch
            res_out = out_ch
            self.up_blocks.append(
                ResidualBlock(res_in, res_out, time_emb_dim=time_mlp_dim)
            )
            if i != len(up_channels) - 1:
                self.upsamples.append(
                    nn.ConvTranspose1d(res_out, res_out, kernel_size=4, stride=2, padding=1)
                )
            in_ch = out_ch

        self.final_conv = nn.Conv1d(up_channels[-1], 1, kernel_size=3, padding=1)

    def forward(self, x_noisy, cond, t):
        x = torch.cat([cond, x_noisy], dim=1)
        t_embed = sinusoidal_time_embedding(t, self.time_emb_dim)
        t_embed = self.time_mlp(t_embed)

        skips = []
        h = self.init_conv(x)

        for i, block in enumerate(self.down_blocks):
            h = block(h, time_emb=t_embed)
            if i != len(self.down_blocks) - 1:
                skips.append(h)
                h = self.downsamples[i](h)

        h = self.attn(h)

        for i, block in enumerate(self.up_blocks):
            skip_feat = skips[-1 - i]
            if h.size(-1) != skip_feat.size(-1):
                h = nn.functional.interpolate(h, size=skip_feat.size(-1), mode="nearest")
            h = torch.cat([h, skip_feat], dim=1)
            h = block(h, time_emb=t_embed)
            if i != len(self.up_blocks) - 1:
                h = self.upsamples[i](h)

        out = self.final_conv(h)
        return out


# =========================================================
# 6. 扩散过程
# =========================================================
class DiffusionScheduler:
    def __init__(self, T, beta_start, beta_end, device):
        self.T = T
        self.device = device

        self.betas = torch.linspace(beta_start, beta_end, T, device=device)
        self.alphas = 1.0 - self.betas
        self.alpha_cum = torch.cumprod(self.alphas, dim=0)

    def diffuse(self, x0, t):
        alpha_bar = self.alpha_cum[t].view(-1, 1, 1)
        noise = torch.randn_like(x0)
        x_t = torch.sqrt(alpha_bar) * x0 + torch.sqrt(1.0 - alpha_bar) * noise
        return x_t, noise

    def sample(self, model, cond, shape):
        model.eval()
        x = torch.randn(shape, device=self.device)

        with torch.no_grad():
            for t_step in reversed(range(self.T)):
                t_cur = torch.full((shape[0],), t_step, device=self.device, dtype=torch.long)
                eps_pred = model(x, cond, t_cur)

                alpha_t = self.alphas[t_step]
                alpha_bar_t = self.alpha_cum[t_step]

                mean = (1.0 / torch.sqrt(alpha_t)) * (
                    x - (1.0 - alpha_t) / torch.sqrt(1.0 - alpha_bar_t) * eps_pred
                )

                if t_step == 0:
                    x = mean
                else:
                    posterior_var = self.betas[t_step] * (1.0 - self.alpha_cum[t_step - 1]) / (1.0 - alpha_bar_t)
                    noise = torch.randn_like(x)
                    x = mean + torch.sqrt(posterior_var) * noise

        return x


# =========================================================
# 7. 评估函数
# =========================================================
def evaluate_noise_mse(model, loader, scheduler, device):
    model.eval()
    total_loss = 0.0
    total_count = 0

    with torch.no_grad():
        for rcg_batch, ecg_batch in loader:
            rcg_batch = rcg_batch.to(device)
            ecg_batch = ecg_batch.to(device)

            bsz = ecg_batch.size(0)
            t = torch.randint(0, scheduler.T, (bsz,), device=device)
            x_t, noise = scheduler.diffuse(ecg_batch, t)
            noise_pred = model(x_t, rcg_batch, t)

            loss = nn.functional.mse_loss(noise_pred, noise, reduction="sum")
            total_loss += loss.item()
            total_count += noise.numel()

    return total_loss / total_count


def evaluate_recon_subset(model, scheduler, rcg_subset, ecg_subset, device):
    model.eval()
    rcg_subset = rcg_subset.to(device)
    ecg_subset = ecg_subset.to(device)

    pred = scheduler.sample(model, rcg_subset, ecg_subset.shape)

    mse = nn.functional.mse_loss(pred, ecg_subset).item()
    mae = nn.functional.l1_loss(pred, ecg_subset).item()

    return mse, mae, pred.detach().cpu()


def evaluate_pearson_loss(model, loader, scheduler, device):
    """
    计算验证集上的皮尔逊损失 (1 - 平均相关系数)
    基于单步扩散的预测 x0
    """
    model.eval()
    total_loss = 0.0
    total_count = 0

    with torch.no_grad():
        for rcg_batch, ecg_batch in loader:
            rcg_batch = rcg_batch.to(device)
            ecg_batch = ecg_batch.to(device)

            bsz = ecg_batch.size(0)
            t = torch.randint(0, scheduler.T, (bsz,), device=device)
            x_t, noise = scheduler.diffuse(ecg_batch, t)
            noise_pred = model(x_t, rcg_batch, t)

            alpha_bar = scheduler.alpha_cum[t].view(-1, 1, 1)
            x0_pred = (x_t - torch.sqrt(1 - alpha_bar) * noise_pred) / (torch.sqrt(alpha_bar) + 1e-8)

            x_centered = x0_pred - x0_pred.mean(dim=-1, keepdim=True)
            y_centered = ecg_batch - ecg_batch.mean(dim=-1, keepdim=True)
            cov = (x_centered * y_centered).sum(dim=-1)
            std_x = torch.sqrt((x_centered ** 2).sum(dim=-1) + 1e-8)
            std_y = torch.sqrt((y_centered ** 2).sum(dim=-1) + 1e-8)
            corr = cov / (std_x * std_y)

            loss = 1 - corr.mean()
            total_loss += loss.item() * bsz
            total_count += bsz

    if total_count == 0:
        return float('nan')
    return total_loss / total_count


def evaluate_combined_loss(model, loader, scheduler, device):
    """
    计算验证集上的组合损失：MSE + 0.1 - 0.1 * 平均相关系数
    """
    model.eval()
    total_loss = 0.0
    total_count = 0

    with torch.no_grad():
        for rcg_batch, ecg_batch in loader:
            rcg_batch = rcg_batch.to(device)
            ecg_batch = ecg_batch.to(device)

            bsz = ecg_batch.size(0)
            t = torch.randint(0, scheduler.T, (bsz,), device=device)
            x_t, noise = scheduler.diffuse(ecg_batch, t)
            noise_pred = model(x_t, rcg_batch, t)

            mse_loss = nn.functional.mse_loss(noise_pred, noise)

            alpha_bar = scheduler.alpha_cum[t].view(-1, 1, 1)
            x0_pred = (x_t - torch.sqrt(1 - alpha_bar) * noise_pred) / (torch.sqrt(alpha_bar) + 1e-8)

            x_centered = x0_pred - x0_pred.mean(dim=-1, keepdim=True)
            y_centered = ecg_batch - ecg_batch.mean(dim=-1, keepdim=True)
            cov = (x_centered * y_centered).sum(dim=-1)
            std_x = torch.sqrt((x_centered ** 2).sum(dim=-1) + 1e-8)
            std_y = torch.sqrt((y_centered ** 2).sum(dim=-1) + 1e-8)
            corr = cov / (std_x * std_y)
            corr_mean = corr.mean()

            loss = mse_loss + 0.1 - 0.1 * corr_mean
            total_loss += loss.item() * bsz
            total_count += bsz

    if total_count == 0:
        return float('nan')
    return total_loss / total_count


# =========================================================
# 8. 保存函数
# =========================================================
def save_json(obj, path):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(obj, f, ensure_ascii=False, indent=2)


def save_checkpoint(path, model, optimizer, epoch, best_eval_mse, cfg, data_meta):
    ckpt = {
        "epoch": epoch,
        "best_val_loss": best_eval_mse,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "model_config": {
            "in_channels": cfg["in_channels"],
            "base_channels": cfg["base_channels"],
            "time_emb_dim": cfg["time_emb_dim"],
            "num_timesteps": cfg["T"]
        },
        "diffusion_config": {
            "T": cfg["T"],
            "beta_start": cfg["beta_start"],
            "beta_end": cfg["beta_end"]
        },
        "data_config": data_meta
    }
    torch.save(ckpt, path)


# =========================================================
# 9. 主函数
# =========================================================
def main():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print("[INFO] Using device:", device)

    # -------------------------
    # 9.1 加载训练集和验证集（不使用测试集）
    # -------------------------
    print("[INFO] Loading train data...")
    train_ecg_list, train_rcg_list, train_prefixes = load_data_from_folder(CFG["train_folder"])

    print("[INFO] Loading eval data...")
    eval_ecg_list, eval_rcg_list, eval_prefixes = load_data_from_folder(CFG["eval_folder"])

    window_len = CFG["signal_length"]
    step = 30

    # 对训练集滑动窗口切分
    train_ecg_samples = []
    train_rcg_samples = []
    for ecg_arr, rcg_arr in zip(train_ecg_list, train_rcg_list):
        if len(ecg_arr) != len(rcg_arr):
            raise ValueError("同一文件中 ECG 和 RCG 长度不一致")
        ecg_windows = sliding_window_cut(ecg_arr, window_len, step)
        rcg_windows = sliding_window_cut(rcg_arr, window_len, step)
        if len(ecg_windows) != len(rcg_windows):
            raise ValueError("滑动窗口后数量不一致")
        train_ecg_samples.extend(ecg_windows)
        train_rcg_samples.extend(rcg_windows)

    # 对验证集滑动窗口切分
    eval_ecg_samples = []
    eval_rcg_samples = []
    for ecg_arr, rcg_arr in zip(eval_ecg_list, eval_rcg_list):
        if len(ecg_arr) != len(rcg_arr):
            raise ValueError("同一文件中 ECG 和 RCG 长度不一致")
        ecg_windows = sliding_window_cut(ecg_arr, window_len, step)
        rcg_windows = sliding_window_cut(rcg_arr, window_len, step)
        if len(ecg_windows) != len(rcg_windows):
            raise ValueError("滑动窗口后数量不一致")
        eval_ecg_samples.extend(ecg_windows)
        eval_rcg_samples.extend(rcg_windows)

    train_ecg = ensure_channel_dim(np.array(train_ecg_samples))
    train_rcg = ensure_channel_dim(np.array(train_rcg_samples))
    eval_ecg = ensure_channel_dim(np.array(eval_ecg_samples))
    eval_rcg = ensure_channel_dim(np.array(eval_rcg_samples))

    assert train_ecg.shape[-1] == window_len
    assert train_rcg.shape[-1] == window_len
    assert eval_ecg.shape[-1] == window_len
    assert eval_rcg.shape[-1] == window_len

    train_ecg_tensor = torch.from_numpy(train_ecg).float()
    train_rcg_tensor = torch.from_numpy(train_rcg).float()
    eval_ecg_tensor = torch.from_numpy(eval_ecg).float()
    eval_rcg_tensor = torch.from_numpy(eval_rcg).float()

    train_dataset = RCGEcgDataset(train_rcg_tensor, train_ecg_tensor)
    eval_dataset = RCGEcgDataset(eval_rcg_tensor, eval_ecg_tensor)

    train_loader = DataLoader(
        train_dataset,
        batch_size=CFG["batch_size"],
        shuffle=True,
        num_workers=CFG["num_workers"]
    )

    eval_loader = DataLoader(
        eval_dataset,
        batch_size=CFG["batch_size"],
        shuffle=False,
        num_workers=CFG["num_workers"]
    )

    print(f"[INFO] Train ECG shape: {train_ecg.shape}")
    print(f"[INFO] Train RCG shape: {train_rcg.shape}")
    print(f"[INFO] Eval ECG shape : {eval_ecg.shape}")
    print(f"[INFO] Eval RCG shape : {eval_rcg.shape}")
    print(f"[INFO] Train prefixes : {train_prefixes[:10]} ... total={len(train_prefixes)}")
    print(f"[INFO] Eval prefixes  : {eval_prefixes[:10]} ... total={len(eval_prefixes)}")

    # -------------------------
    # 9.2 保存配置
    # -------------------------
    data_meta = {
        "train_folder": CFG["train_folder"],
        "eval_folder": CFG["eval_folder"],
        "signal_length": CFG["signal_length"],
        "rcg_channels": CFG["rcg_channels"],
        "ecg_channels": CFG["ecg_channels"],
        "raw_processing": {
            "source_mat_range": "1.mat ~ 90.mat",
            "best_rcg_channel_rule": "choose channel with highest Pearson correlation to ECG",
            "window_length": 640,
            "step": 30,
            "note": "no test set is used at all; only train + val"
        }
    }

    save_json(CFG, ART_DIR / "train_config.json")
    save_json(data_meta, ART_DIR / "data_config.json")

    # -------------------------
    # 9.3 模型、优化器、调度器
    # -------------------------
    model = ConditionalDiffusionModel(
        in_channels=CFG["in_channels"],
        base_channels=CFG["base_channels"],
        time_emb_dim=CFG["time_emb_dim"],
        num_timesteps=CFG["T"]
    ).to(device)

    optimizer = optim.Adam(model.parameters(), lr=CFG["learning_rate"])
    scheduler = DiffusionScheduler(
        T=CFG["T"],
        beta_start=CFG["beta_start"],
        beta_end=CFG["beta_end"],
        device=device
    )

    # ========== 加载已有 checkpoint 继续训练 ==========
    resume_path = ART_DIR / "last_model.pt"
    start_epoch = 1
    best_val_combined = float("inf")

    if resume_path.exists():
        print(f"[INFO] Loading checkpoint from {resume_path}")
        ckpt = torch.load(resume_path, map_location=device)
        model.load_state_dict(ckpt["model_state_dict"])
        optimizer.load_state_dict(ckpt["optimizer_state_dict"])
        start_epoch = ckpt["epoch"] + 1
        best_val_combined = ckpt.get("best_val_loss", float("inf"))
        print(f"[INFO] Resumed from epoch {ckpt['epoch']}, best_val_combined={best_val_combined:.6f}")
    else:
        print("[INFO] No checkpoint found, starting from scratch.")

    # 固定小样本子集用于重建评估
    eval_subset_size = min(CFG["eval_subset_size"], len(eval_dataset))
    eval_subset_rcg = eval_rcg_tensor[:eval_subset_size].clone()
    eval_subset_ecg = eval_ecg_tensor[:eval_subset_size].clone()

    print(f"[INFO] Fixed eval subset size: {eval_subset_size}")

    # -------------------------
    # 9.4 训练循环
    # -------------------------
    history = []

    total_train_start = time.time()
    total_train_time = 0.0   # 所有 epoch 的纯训练时间累加
    total_epoch_time = 0.0   # 所有 epoch 的总时间（训练+验证）累加

    for epoch in range(start_epoch, CFG["num_epochs"] + 1):

        epoch_start = time.time()
        model.train()

        running_loss = 0.0
        running_count = 0

        # =========================
        # Training
        # =========================
        train_start = time.time()

        for rcg_batch, ecg_batch in train_loader:

            rcg_batch = rcg_batch.to(device)
            ecg_batch = ecg_batch.to(device)

            bsz = ecg_batch.size(0)

            t = torch.randint(0, CFG["T"], (bsz,), device=device)

            x_t, noise = scheduler.diffuse(ecg_batch, t)

            noise_pred = model(x_t, rcg_batch, t)

            # Loss: MSE + 0.1 - 0.1*corr
            mse_loss = nn.functional.mse_loss(noise_pred, noise)

            alpha_bar = scheduler.alpha_cum[t].view(-1, 1, 1)
            x0_pred = (x_t - torch.sqrt(1 - alpha_bar) * noise_pred) / (torch.sqrt(alpha_bar) + 1e-8)

            x_centered = x0_pred - x0_pred.mean(dim=-1, keepdim=True)
            y_centered = ecg_batch - ecg_batch.mean(dim=-1, keepdim=True)

            cov = (x_centered * y_centered).sum(dim=-1)
            std_x = torch.sqrt((x_centered ** 2).sum(dim=-1) + 1e-8)
            std_y = torch.sqrt((y_centered ** 2).sum(dim=-1) + 1e-8)
            corr = cov / (std_x * std_y)
            corr_mean = corr.mean()

            loss = mse_loss + 0.1 - 0.1 * corr_mean

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            running_loss += loss.item() * bsz
            running_count += bsz

        train_time = time.time() - train_start
        train_loss = running_loss / max(running_count, 1)

        # =========================
        # Validation（仅使用验证集）
        # =========================
        val_start = time.time()

        eval_noise_mse = evaluate_noise_mse(model, eval_loader, scheduler, device)

        val_combined_loss = evaluate_combined_loss(model, eval_loader, scheduler, device)

        recon_mse, recon_mae, _ = evaluate_recon_subset(
            model, scheduler, eval_subset_rcg, eval_subset_ecg, device
        )

        val_time = time.time() - val_start
        epoch_time = time.time() - epoch_start

        total_train_time += train_time
        total_epoch_time += epoch_time

        epoch_info = {
            "epoch": epoch,
            "train_loss": float(train_loss),
            "val_combined_loss": float(val_combined_loss),
            "eval_noise_mse": float(eval_noise_mse),
            "eval_recon_mse_subset": float(recon_mse),
            "eval_recon_mae_subset": float(recon_mae),
            "train_time_sec": float(train_time),
            "val_time_sec": float(val_time),
            "epoch_time_sec": float(epoch_time),
        }
        history.append(epoch_info)

        print(
            f"[Epoch {epoch:03d}] "
            f"train={train_loss:.6f} | "
            f"val_combined={val_combined_loss:.6f} | "
            f"eval_noise_mse={eval_noise_mse:.6f} | "
            f"recon_mse={recon_mse:.6f} | "
            f"recon_mae={recon_mae:.6f} | "
            f"train_time={format_seconds(train_time)} | "
            f"val_time={format_seconds(val_time)} | "
            f"epoch_time={format_seconds(epoch_time)}"
        )

        # =================================================
        # 保存最佳模型（基于验证集组合损失，越小越好）
        # =================================================
        if val_combined_loss < best_val_combined:
            best_val_combined = val_combined_loss

            save_checkpoint(
                ART_DIR / "best_model.pt",
                model=model,
                optimizer=optimizer,
                epoch=epoch,
                best_eval_mse=best_val_combined,
                cfg=CFG,
                data_meta=data_meta
            )
            print(
                "[INFO] Best model updated "
                f"epoch={epoch}, val_combined={best_val_combined:.6f}"
            )

        # =================================================
        # 保存最后模型
        # =================================================
        save_checkpoint(
            ART_DIR / "last_model.pt",
            model=model,
            optimizer=optimizer,
            epoch=epoch,
            best_eval_mse=best_val_combined,
            cfg=CFG,
            data_meta=data_meta
        )

        # 定期保存 history
        if epoch % CFG["save_every"] == 0:
            save_json(history, ART_DIR / "train_history.json")

    save_json(history, ART_DIR / "train_history.json")

    total_wall_time = time.time() - total_train_start

    print("[INFO] Training complete.")
    print(f"[INFO] Best validation combined loss: {best_val_combined:.6f}")
    print(f"[INFO] Total training time (sum of pure train time): {format_seconds(total_train_time)}")
    print(f"[INFO] Total epoch time (sum of train+val time)    : {format_seconds(total_epoch_time)}")
    print(f"[INFO] Total wall time (from start to end)          : {format_seconds(total_wall_time)}")
    print(f"[INFO] Artifacts saved to: {ART_DIR.resolve()}")


if __name__ == "__main__":
    main()
```

```python
import os
import json
import math
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn


# =========================================================
# 0. 路径与单样本配置
# =========================================================
ART_DIR = Path("/home/ubuntu-user/Desktop/code-9-8/our-9-27-raw")
CKPT_PATH = ART_DIR / "last_model.pt"
TRAIN_CFG_PATH = ART_DIR / "train_config.json"
DATA_CFG_PATH = ART_DIR / "data_config.json"

# 指定单样本：第几个文件前缀、该文件里的第几条片段
TARGET_PREFIX = 10
TARGET_INDEX = 0

SAVE_DIR = ART_DIR / "single_infer_last"
SAVE_DIR.mkdir(parents=True, exist_ok=True)


def sliding_window_cut(data, window_len, step=30):
    """
    将一维数据 data 切成长度为 window_len 的片段，步长为 step。
    返回所有片段的列表。
    """
    if len(data) < window_len:
        return []
    indices = range(0, len(data) - window_len + 1, step)
    return [data[i:i+window_len] for i in indices]

# =========================================================
# 1. 与训练一致的模型定义
# =========================================================
def sinusoidal_time_embedding(timesteps, embed_dim):
    device = timesteps.device
    half_dim = embed_dim // 2
    exp_range = torch.arange(half_dim, device=device, dtype=torch.float32)
    exp_range = -math.log(10000.0) * exp_range / float(half_dim)
    sinusoid_inp = timesteps.float()[:, None] * torch.exp(exp_range)[None, :]
    emb = torch.cat([torch.sin(sinusoid_inp), torch.cos(sinusoid_inp)], dim=1)
    return emb


class SqueezeExcitation(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        reduced_channels = max(1, channels // reduction)
        self.avg_pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, reduced_channels),
            nn.SiLU(),
            nn.Linear(reduced_channels, channels),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _ = x.size()
        y = self.avg_pool(x).view(b, c)
        y = self.fc(y).view(b, c, 1)
        return x * y


class SelfAttention1D(nn.Module):
    def __init__(self, channels):
        super().__init__()

        if channels < 8:
            num_heads = 1
        elif channels % 8 == 0:
            num_heads = 8
        else:
            num_heads = max(1, channels // 2)

        if channels >= 8 and channels % 8 == 0:
            groups = 8
        elif channels >= 4 and channels % 4 == 0:
            groups = 4
        elif channels >= 2 and channels % 2 == 0:
            groups = 2
        else:
            groups = 1

        self.norm = nn.GroupNorm(num_groups=groups, num_channels=channels)
        self.attn = nn.MultiheadAttention(
            embed_dim=channels,
            num_heads=num_heads,
            batch_first=True
        )

    def forward(self, x):
        h = self.norm(x)
        h = h.permute(0, 2, 1)
        attn_out, _ = self.attn(h, h, h)
        attn_out = attn_out.permute(0, 2, 1)
        return x + attn_out


class ResidualBlock(nn.Module):
    def __init__(self, in_channels, out_channels, time_emb_dim=None):
        super().__init__()

        def make_norm(channels):
            if channels >= 8 and channels % 8 == 0:
                groups = 8
            elif channels >= 4 and channels % 4 == 0:
                groups = 4
            elif channels >= 2 and channels % 2 == 0:
                groups = 2
            else:
                groups = 1
            return nn.GroupNorm(num_groups=groups, num_channels=channels)

        self.norm1 = make_norm(in_channels)
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size=3, padding=1)

        self.time_fc = nn.Linear(time_emb_dim, out_channels) if time_emb_dim is not None else None

        self.norm2 = make_norm(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size=3, padding=1)

        self.se = SqueezeExcitation(out_channels)
        self.shortcut = nn.Conv1d(in_channels, out_channels, kernel_size=1) if in_channels != out_channels else None
        self.act = nn.SiLU()

    def forward(self, x, time_emb=None):
        h = self.norm1(x)
        h = self.act(h)
        h = self.conv1(h)

        if self.time_fc is not None and time_emb is not None:
            time_bias = self.time_fc(time_emb)
            h = h + time_bias[..., None]

        h = self.norm2(h)
        h = self.act(h)
        h = self.conv2(h)

        h = self.se(h)

        x_short = self.shortcut(x) if self.shortcut is not None else x
        return x_short + h


class ConditionalDiffusionModel(nn.Module):
    def __init__(self, in_channels=2, base_channels=64, time_emb_dim=128, num_timesteps=100):
        super().__init__()
        self.in_channels = in_channels
        self.base_channels = base_channels
        self.time_emb_dim = time_emb_dim
        self.num_timesteps = num_timesteps

        time_mlp_dim = base_channels * 8
        self.time_mlp = nn.Sequential(
            nn.Linear(time_emb_dim, time_mlp_dim),
            nn.SiLU(),
            nn.Linear(time_mlp_dim, time_mlp_dim)
        )

        self.init_conv = nn.Conv1d(in_channels, base_channels, kernel_size=3, padding=1)

        self.down_blocks = nn.ModuleList()
        self.downsamples = nn.ModuleList()

        down_channels = [
            base_channels,
            base_channels * 2,
            base_channels * 4,
            base_channels * 8,
            base_channels * 8,
            base_channels * 8
        ]

        in_ch = base_channels
        for i, out_ch in enumerate(down_channels):
            self.down_blocks.append(
                ResidualBlock(in_ch, out_ch, time_emb_dim=time_mlp_dim)
            )
            if i != len(down_channels) - 1:
                self.downsamples.append(
                    nn.Conv1d(out_ch, out_ch, kernel_size=4, stride=2, padding=1)
                )
            in_ch = out_ch

        self.attn = SelfAttention1D(down_channels[-1])

        self.up_blocks = nn.ModuleList()
        self.upsamples = nn.ModuleList()

        up_channels = [
            base_channels * 8,
            base_channels * 4,
            base_channels * 2,
            base_channels,
            base_channels
        ]

        in_ch = down_channels[-1]
        for i, out_ch in enumerate(up_channels):
            skip_ch = down_channels[-2 - i]
            res_in = in_ch + skip_ch
            self.up_blocks.append(
                ResidualBlock(res_in, out_ch, time_emb_dim=time_mlp_dim)
            )
            if i != len(up_channels) - 1:
                self.upsamples.append(
                    nn.ConvTranspose1d(out_ch, out_ch, kernel_size=4, stride=2, padding=1)
                )
            in_ch = out_ch

        self.final_conv = nn.Conv1d(up_channels[-1], 1, kernel_size=3, padding=1)

    def forward(self, x_noisy, cond, t):
        x = torch.cat([cond, x_noisy], dim=1)
        t_embed = sinusoidal_time_embedding(t, self.time_emb_dim)
        t_embed = self.time_mlp(t_embed)

        skips = []
        h = self.init_conv(x)

        for i, block in enumerate(self.down_blocks):
            h = block(h, time_emb=t_embed)
            if i != len(self.down_blocks) - 1:
                skips.append(h)
                h = self.downsamples[i](h)

        h = self.attn(h)

        for i, block in enumerate(self.up_blocks):
            skip_feat = skips[-1 - i]
            if h.size(-1) != skip_feat.size(-1):
                h = nn.functional.interpolate(h, size=skip_feat.size(-1), mode="nearest")
            h = torch.cat([h, skip_feat], dim=1)
            h = block(h, time_emb=t_embed)
            if i != len(self.up_blocks) - 1:
                h = self.upsamples[i](h)

        return self.final_conv(h)


class DiffusionScheduler:
    def __init__(self, T, beta_start, beta_end, device):
        self.T = T
        self.device = device
        self.betas = torch.linspace(beta_start, beta_end, T, device=device)
        self.alphas = 1.0 - self.betas
        self.alpha_cum = torch.cumprod(self.alphas, dim=0)

    def sample(self, model, cond, shape):
        model.eval()
        x = torch.randn(shape, device=self.device)

        with torch.no_grad():
            for t_step in reversed(range(self.T)):
                t_cur = torch.full((shape[0],), t_step, device=self.device, dtype=torch.long)
                eps_pred = model(x, cond, t_cur)

                alpha_t = self.alphas[t_step]
                alpha_bar_t = self.alpha_cum[t_step]

                mean = (1.0 / torch.sqrt(alpha_t)) * (
                    x - (1.0 - alpha_t) / torch.sqrt(1.0 - alpha_bar_t) * eps_pred
                )

                if t_step == 0:
                    x = mean
                else:
                    posterior_var = self.betas[t_step] * (1.0 - self.alpha_cum[t_step - 1]) / (1.0 - alpha_bar_t)
                    noise = torch.randn_like(x)
                    x = mean + torch.sqrt(posterior_var) * noise
        return x


# =========================================================
# 2. 读单条样本
# =========================================================
def load_single_sample(eval_folder, prefix, index, window_len=640, step=30):
    csv_path = os.path.join(eval_folder, f"{prefix}.csv")
    if not os.path.exists(csv_path):
        raise FileNotFoundError(f"找不到 {csv_path}")

    try:
        data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
    except Exception as e:
        raise RuntimeError(f"读取 {csv_path} 失败: {e}")

    if 'ECG' not in data.dtype.names or 'RCG' not in data.dtype.names:
        raise ValueError(f"{csv_path} 中缺少 'ECG' 或 'RCG' 列")

    ecg_all = data['ECG']
    rcg_all = data['RCG']

    if len(ecg_all) != len(rcg_all):
        raise ValueError(f"{csv_path} 中 ECG 和 RCG 长度不一致")

    # 滑动窗口切割
    ecg_windows = sliding_window_cut(ecg_all, window_len, step)
    rcg_windows = sliding_window_cut(rcg_all, window_len, step)

    if len(ecg_windows) == 0:
        raise RuntimeError(f"{csv_path} 长度不足 {window_len}，无法切出窗口")

    if index < 0 or index >= len(ecg_windows):
        raise IndexError(f"TARGET_INDEX={index} 超出范围，该文件共有 {len(ecg_windows)} 个窗口")

    rcg = rcg_windows[index]
    ecg = ecg_windows[index]

    # 转为 (1, L) 形状
    if rcg.ndim == 1:
        rcg = rcg[None, :]
    if ecg.ndim == 1:
        ecg = ecg[None, :]

    return rcg.astype(np.float32), ecg.astype(np.float32)


# =========================================================
# 3. 主流程
# =========================================================
def main():
    assert CKPT_PATH.exists(), f"找不到 checkpoint: {CKPT_PATH}"
    assert TRAIN_CFG_PATH.exists(), f"找不到配置文件: {TRAIN_CFG_PATH}"
    assert DATA_CFG_PATH.exists(), f"找不到配置文件: {DATA_CFG_PATH}"

    with open(TRAIN_CFG_PATH, "r", encoding="utf-8") as f:
        train_cfg = json.load(f)

    with open(DATA_CFG_PATH, "r", encoding="utf-8") as f:
        data_cfg = json.load(f)

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print("[INFO] Using device:", device)

    ckpt = torch.load(CKPT_PATH, map_location=device)
    model_cfg = ckpt["model_config"]
    diffusion_cfg = ckpt["diffusion_config"]

    print("[INFO] Loading checkpoint from:", CKPT_PATH)
    print("[INFO] model_config:", model_cfg)
    print("[INFO] diffusion_config:", diffusion_cfg)
    print("[INFO] checkpoint epoch:", ckpt.get("epoch", "unknown"))

    model = ConditionalDiffusionModel(
        in_channels=model_cfg["in_channels"],
        base_channels=model_cfg["base_channels"],
        time_emb_dim=model_cfg["time_emb_dim"],
        num_timesteps=model_cfg["num_timesteps"]
    ).to(device)

    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()
    print("[INFO] Model reloaded successfully.")

    scheduler = DiffusionScheduler(
        T=diffusion_cfg["T"],
        beta_start=diffusion_cfg["beta_start"],
        beta_end=diffusion_cfg["beta_end"],
        device=device
    )

    eval_folder = "/home/ubuntu-user/Desktop/code-9-8/spilt-9-27-raw/test"
    signal_length = data_cfg["signal_length"]
    # 从配置中获取步长，若无则默认 30（与训练一致）
    #step = data_cfg.get("raw_processing", {}).get("step", 30)
    step = 640
    print("[INFO] Loading single sample from:", eval_folder)

        # ========== 批量处理配置 ==========
    # 自动获取所有 CSV 文件的数字前缀
    all_csv_files = [f for f in os.listdir(eval_folder) if f.endswith(".csv")]
    prefixes = []
    for f in all_csv_files:
        try:
            pref = int(f.split(".")[0])
            prefixes.append(pref)
        except ValueError:
            continue
    prefixes = sorted(prefixes)
    print(f"[INFO] 找到 {len(prefixes)} 个 CSV 文件: {prefixes[:10]}{'...' if len(prefixes)>10 else ''}")

    # 每个文件最多取多少个窗口（None 表示全取）
    max_windows_per_file = 100          # 改为 None 则取全部
    total_sample_limit = 1000          # 总样本数上限，设为 None 则不限制

    sample_list = []   # (prefix, index)
    for pref in prefixes:
        csv_path = os.path.join(eval_folder, f"{pref}.csv")
        try:
            data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
        except:
            continue
        if 'ECG' not in data.dtype.names or 'RCG' not in data.dtype.names:
            continue
        ecg_all = data['ECG']
        rcg_all = data['RCG']
        if len(ecg_all) != len(rcg_all):
            continue
        # 计算窗口总数
        num_windows = max(0, (len(ecg_all) - signal_length) // step + 1)
        if num_windows <= 0:
            continue
        # 决定取哪些窗口索引
        if max_windows_per_file is not None:
            indices = range(min(max_windows_per_file, num_windows))
        else:
            indices = range(num_windows)
        for idx in indices:
            sample_list.append((pref, idx))
            if total_sample_limit is not None and len(sample_list) >= total_sample_limit:
                break
        if total_sample_limit is not None and len(sample_list) >= total_sample_limit:
            break

    print(f"[INFO] 共生成 {len(sample_list)} 个待处理样本")

    # ========== 存储结果 ==========
    results = []
    rcg_list, ecg_true_list, ecg_pred_list = [], [], []

    for pref, idx in sample_list:
        try:
            rcg_np, ecg_np = load_single_sample(eval_folder, pref, idx, signal_length, step)
        except (FileNotFoundError, IndexError, RuntimeError) as e:
            print(f"[WARN] 跳过 ({pref}, {idx})：{e}")
            continue

        rcg = torch.from_numpy(rcg_np).unsqueeze(0).to(device)
        ecg_true = torch.from_numpy(ecg_np).unsqueeze(0).to(device)

        ecg_pred = scheduler.sample(model, rcg, ecg_true.shape)

        rcg_arr = rcg[0, 0].cpu().numpy()
        ecg_true_arr = ecg_true[0, 0].cpu().numpy()
        ecg_pred_arr = ecg_pred[0, 0].cpu().numpy()

        mse = np.mean((ecg_true_arr - ecg_pred_arr) ** 2)
        mae = np.mean(np.abs(ecg_true_arr - ecg_pred_arr))
        corr = np.corrcoef(ecg_true_arr, ecg_pred_arr)[0, 1]

        results.append({
            "prefix": int(pref),
            "index": int(idx),
            "mse": float(mse),
            "mae": float(mae),
            "correlation": float(corr)
        })

        if len(results) <= 20:   # 只保留前20个样本用于绘图
            rcg_list.append(rcg_arr)
            ecg_true_list.append(ecg_true_arr)
            ecg_pred_list.append(ecg_pred_arr)

        print(f"[INFO] ({pref}, {idx}) MSE={mse:.6f}, MAE={mae:.6f}, Corr={corr:.6f}")

    # ========== 汇总统计（同上，保持不变） ==========
    if results:
        corr_values = [r["correlation"] for r in results]
        mse_values = [r["mse"] for r in results]
        mae_values = [r["mae"] for r in results]

        summary_stats = {
            "num_samples": len(results),
            "correlation_mean": float(np.mean(corr_values)),
            "correlation_std": float(np.std(corr_values)),
            "correlation_min": float(np.min(corr_values)),
            "correlation_max": float(np.max(corr_values)),
            "mse_mean": float(np.mean(mse_values)),
            "mse_std": float(np.std(mse_values)),
            "mae_mean": float(np.mean(mae_values)),
            "mae_std": float(np.std(mae_values)),
        }
        print("\n========== 汇总统计 ==========")
        for k, v in summary_stats.items():
            print(f"{k}: {v:.6f}")

        full_results = {
            "samples": results,
            "summary": summary_stats,
            "config": {
                "checkpoint": str(CKPT_PATH),
                "max_windows_per_file": max_windows_per_file,
                "total_sample_limit": total_sample_limit,
                "signal_length": signal_length,
                "step": step
            }
        }
        with open(SAVE_DIR / "batch_results.json", "w", encoding="utf-8") as f:
            json.dump(full_results, f, indent=2)

        # 绘制前几个样本
                # 绘制前几个样本（仅显示，不保存）
        num_plot = min(len(rcg_list), 4)
        if num_plot > 0:
            fig, axes = plt.subplots(num_plot, 3, figsize=(15, 4*num_plot))
            if num_plot == 1:
                axes = axes.reshape(1, -1)
            for i in range(num_plot):
                axes[i, 0].plot(rcg_list[i])
                axes[i, 0].set_title(f"RCG (sample {i+1})")
                axes[i, 1].plot(ecg_true_list[i])
                axes[i, 1].set_title("True ECG")
                axes[i, 2].plot(ecg_pred_list[i])
                axes[i, 2].set_title(f"Pred ECG (corr={results[i]['correlation']:.3f})")
            plt.tight_layout()
            plt.show()   # 直接显示，不保存文件
            print("[INFO] 已显示样本对比图")
if __name__ == "__main__":
    main()
```

```python
import os
import json
import math
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import random
def set_seed(seed=0):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(0)
# =========================================================
# 0. 路径与配置
# =========================================================
ART_DIR = Path("/home/ubuntu-user/Desktop/code-9-8/our-9-27-raw")
CKPT_PATH = ART_DIR / "last_model.pt"
TRAIN_CFG_PATH = ART_DIR / "train_config.json"
DATA_CFG_PATH = ART_DIR / "data_config.json"

# 评估时从信号的第几个点开始计算指标（跳过开头）
EVAL_START = 100   # 可修改，与训练时的 cut_head 一致
SAMPLING_RATE = 200
SAVE_DIR = ART_DIR / "single_infer_last"
SAVE_DIR.mkdir(parents=True, exist_ok=True)


def sliding_window_cut(data, window_len, step=30):
    """
    将一维数据 data 切成长度为 window_len 的片段，步长为 step。
    返回所有片段的列表。
    """
    if len(data) < window_len:
        return []
    indices = range(0, len(data) - window_len + 1, step)
    return [data[i:i+window_len] for i in indices]


# =========================================================
# 1. 与训练一致的模型定义
# =========================================================
def sinusoidal_time_embedding(timesteps, embed_dim):
    device = timesteps.device
    half_dim = embed_dim // 2
    exp_range = torch.arange(half_dim, device=device, dtype=torch.float32)
    exp_range = -math.log(10000.0) * exp_range / float(half_dim)
    sinusoid_inp = timesteps.float()[:, None] * torch.exp(exp_range)[None, :]
    emb = torch.cat([torch.sin(sinusoid_inp), torch.cos(sinusoid_inp)], dim=1)
    return emb


class SqueezeExcitation(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        reduced_channels = max(1, channels // reduction)
        self.avg_pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, reduced_channels),
            nn.SiLU(),
            nn.Linear(reduced_channels, channels),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _ = x.size()
        y = self.avg_pool(x).view(b, c)
        y = self.fc(y).view(b, c, 1)
        return x * y


class SelfAttention1D(nn.Module):
    def __init__(self, channels):
        super().__init__()

        if channels < 8:
            num_heads = 1
        elif channels % 8 == 0:
            num_heads = 8
        else:
            num_heads = max(1, channels // 2)

        if channels >= 8 and channels % 8 == 0:
            groups = 8
        elif channels >= 4 and channels % 4 == 0:
            groups = 4
        elif channels >= 2 and channels % 2 == 0:
            groups = 2
        else:
            groups = 1

        self.norm = nn.GroupNorm(num_groups=groups, num_channels=channels)
        self.attn = nn.MultiheadAttention(
            embed_dim=channels,
            num_heads=num_heads,
            batch_first=True
        )

    def forward(self, x):
        h = self.norm(x)
        h = h.permute(0, 2, 1)
        attn_out, _ = self.attn(h, h, h)
        attn_out = attn_out.permute(0, 2, 1)
        return x + attn_out


class ResidualBlock(nn.Module):
    def __init__(self, in_channels, out_channels, time_emb_dim=None):
        super().__init__()

        def make_norm(channels):
            if channels >= 8 and channels % 8 == 0:
                groups = 8
            elif channels >= 4 and channels % 4 == 0:
                groups = 4
            elif channels >= 2 and channels % 2 == 0:
                groups = 2
            else:
                groups = 1
            return nn.GroupNorm(num_groups=groups, num_channels=channels)

        self.norm1 = make_norm(in_channels)
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size=3, padding=1)

        self.time_fc = nn.Linear(time_emb_dim, out_channels) if time_emb_dim is not None else None

        self.norm2 = make_norm(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size=3, padding=1)

        self.se = SqueezeExcitation(out_channels)
        self.shortcut = nn.Conv1d(in_channels, out_channels, kernel_size=1) if in_channels != out_channels else None
        self.act = nn.SiLU()

    def forward(self, x, time_emb=None):
        h = self.norm1(x)
        h = self.act(h)
        h = self.conv1(h)

        if self.time_fc is not None and time_emb is not None:
            time_bias = self.time_fc(time_emb)
            h = h + time_bias[..., None]

        h = self.norm2(h)
        h = self.act(h)
        h = self.conv2(h)

        h = self.se(h)

        x_short = self.shortcut(x) if self.shortcut is not None else x
        return x_short + h


class ConditionalDiffusionModel(nn.Module):
    def __init__(self, in_channels=2, base_channels=64, time_emb_dim=128, num_timesteps=100):
        super().__init__()
        self.in_channels = in_channels
        self.base_channels = base_channels
        self.time_emb_dim = time_emb_dim
        self.num_timesteps = num_timesteps

        time_mlp_dim = base_channels * 8
        self.time_mlp = nn.Sequential(
            nn.Linear(time_emb_dim, time_mlp_dim),
            nn.SiLU(),
            nn.Linear(time_mlp_dim, time_mlp_dim)
        )

        self.init_conv = nn.Conv1d(in_channels, base_channels, kernel_size=3, padding=1)

        self.down_blocks = nn.ModuleList()
        self.downsamples = nn.ModuleList()

        down_channels = [
            base_channels,
            base_channels * 2,
            base_channels * 4,
            base_channels * 8,
            base_channels * 8,
            base_channels * 8
        ]

        in_ch = base_channels
        for i, out_ch in enumerate(down_channels):
            self.down_blocks.append(
                ResidualBlock(in_ch, out_ch, time_emb_dim=time_mlp_dim)
            )
            if i != len(down_channels) - 1:
                self.downsamples.append(
                    nn.Conv1d(out_ch, out_ch, kernel_size=4, stride=2, padding=1)
                )
            in_ch = out_ch

        self.attn = SelfAttention1D(down_channels[-1])

        self.up_blocks = nn.ModuleList()
        self.upsamples = nn.ModuleList()

        up_channels = [
            base_channels * 8,
            base_channels * 4,
            base_channels * 2,
            base_channels,
            base_channels
        ]

        in_ch = down_channels[-1]
        for i, out_ch in enumerate(up_channels):
            skip_ch = down_channels[-2 - i]
            res_in = in_ch + skip_ch
            self.up_blocks.append(
                ResidualBlock(res_in, out_ch, time_emb_dim=time_mlp_dim)
            )
            if i != len(up_channels) - 1:
                self.upsamples.append(
                    nn.ConvTranspose1d(out_ch, out_ch, kernel_size=4, stride=2, padding=1)
                )
            in_ch = out_ch

        self.final_conv = nn.Conv1d(up_channels[-1], 1, kernel_size=3, padding=1)

    def forward(self, x_noisy, cond, t):
        x = torch.cat([cond, x_noisy], dim=1)
        t_embed = sinusoidal_time_embedding(t, self.time_emb_dim)
        t_embed = self.time_mlp(t_embed)

        skips = []
        h = self.init_conv(x)

        for i, block in enumerate(self.down_blocks):
            h = block(h, time_emb=t_embed)
            if i != len(self.down_blocks) - 1:
                skips.append(h)
                h = self.downsamples[i](h)

        h = self.attn(h)

        for i, block in enumerate(self.up_blocks):
            skip_feat = skips[-1 - i]
            if h.size(-1) != skip_feat.size(-1):
                h = nn.functional.interpolate(h, size=skip_feat.size(-1), mode="nearest")
            h = torch.cat([h, skip_feat], dim=1)
            h = block(h, time_emb=t_embed)
            if i != len(self.up_blocks) - 1:
                h = self.upsamples[i](h)

        return self.final_conv(h)


class DiffusionScheduler:
    def __init__(self, T, beta_start, beta_end, device):
        self.T = T
        self.device = device
        self.betas = torch.linspace(beta_start, beta_end, T, device=device)
        self.alphas = 1.0 - self.betas
        self.alpha_cum = torch.cumprod(self.alphas, dim=0)

    def sample(self, model, cond, shape):
        model.eval()
        x = torch.randn(shape, device=self.device)

        with torch.no_grad():
            for t_step in reversed(range(self.T)):
                t_cur = torch.full((shape[0],), t_step, device=self.device, dtype=torch.long)
                eps_pred = model(x, cond, t_cur)

                alpha_t = self.alphas[t_step]
                alpha_bar_t = self.alpha_cum[t_step]

                mean = (1.0 / torch.sqrt(alpha_t)) * (
                    x - (1.0 - alpha_t) / torch.sqrt(1.0 - alpha_bar_t) * eps_pred
                )

                if t_step == 0:
                    x = mean
                else:
                    posterior_var = self.betas[t_step] * (1.0 - self.alpha_cum[t_step - 1]) / (1.0 - alpha_bar_t)
                    noise = torch.randn_like(x)
                    x = mean + torch.sqrt(posterior_var) * noise
        return x


# =========================================================
# 2. 读单条样本（支持裁剪开头）
# =========================================================
# =========================================================
# 2. 读单条样本（返回条件 RCG，目标 ECG）
# =========================================================
# 2. 读单条样本（返回条件 RCG，目标 ECG）
# =========================================================
def load_single_sample(eval_folder, prefix, index, window_len=640, step=30, cut_head=100):
    csv_path = os.path.join(eval_folder, f"{prefix}.csv")
    if not os.path.exists(csv_path):
        raise FileNotFoundError(f"找不到 {csv_path}")

    try:
        data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
    except Exception as e:
        raise RuntimeError(f"读取 {csv_path} 失败: {e}")

    if 'ECG' not in data.dtype.names or 'RCG' not in data.dtype.names:
        raise ValueError(f"{csv_path} 中缺少 'ECG' 或 'RCG' 列")

    ecg_all = data['ECG']
    rcg_all = data['RCG']

    if len(ecg_all) != len(rcg_all):
        raise ValueError(f"{csv_path} 中 ECG 和 RCG 长度不一致")

    if len(ecg_all) > cut_head:
        ecg_all = ecg_all[cut_head:]
        rcg_all = rcg_all[cut_head:]
    else:
        raise RuntimeError(f"{csv_path} 长度 {len(ecg_all)} <= {cut_head}，无法裁剪")

    ecg_windows = sliding_window_cut(ecg_all, window_len, step)
    rcg_windows = sliding_window_cut(rcg_all, window_len, step)

    if len(ecg_windows) == 0:
        raise RuntimeError(f"{csv_path} 裁剪后长度不足 {window_len}，无法切出窗口")

    if index < 0 or index >= len(ecg_windows):
        raise IndexError(f"TARGET_INDEX={index} 超出范围，该文件共有 {len(ecg_windows)} 个窗口")

    rcg = rcg_windows[index]
    ecg = ecg_windows[index]

    if rcg.ndim == 1:
        rcg = rcg[None, :]
    if ecg.ndim == 1:
        ecg = ecg[None, :]

    # 返回 (条件 RCG, 目标 ECG)
    return rcg.astype(np.float32), ecg.astype(np.float32)

def get_peak_freq(signal, fs):
    """
    计算信号幅度谱峰值对应的频率。
    参数：
        signal: 一维数组
        fs: 采样率 (Hz)
    返回：
        峰值频率 (单位 Hz)
    """
    n = len(signal)
    if n < 2:
        return 0.0
    fft_vals = np.fft.rfft(signal)
    magnitudes = np.abs(fft_vals)
    magnitudes[0] = 0.0          # 忽略直流分量
    peak_idx = np.argmax(magnitudes)
    freqs = np.fft.rfftfreq(n, d=1/fs)
    return freqs[peak_idx]

# =========================================================
# 3. 主流程
# =========================================================
def main():
    assert CKPT_PATH.exists(), f"找不到 checkpoint: {CKPT_PATH}"
    assert TRAIN_CFG_PATH.exists(), f"找不到配置文件: {TRAIN_CFG_PATH}"
    assert DATA_CFG_PATH.exists(), f"找不到配置文件: {DATA_CFG_PATH}"

    save_folder = Path("/home/ubuntu-user/Desktop/code-9-8/generate-9-27")
    save_folder.mkdir(parents=True, exist_ok=True)
    print(f"[INFO] 生成的数据将保存到: {save_folder}")

    with open(TRAIN_CFG_PATH, "r", encoding="utf-8") as f:
        train_cfg = json.load(f)

    with open(DATA_CFG_PATH, "r", encoding="utf-8") as f:
        data_cfg = json.load(f)

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print("[INFO] Using device:", device)

    ckpt = torch.load(CKPT_PATH, map_location=device)
    model_cfg = ckpt["model_config"]
    diffusion_cfg = ckpt["diffusion_config"]

    print("[INFO] Loading checkpoint from:", CKPT_PATH)
    print("[INFO] model_config:", model_cfg)
    print("[INFO] diffusion_config:", diffusion_cfg)
    print("[INFO] checkpoint epoch:", ckpt.get("epoch", "unknown"))

    model = ConditionalDiffusionModel(
        in_channels=model_cfg["in_channels"],
        base_channels=model_cfg["base_channels"],
        time_emb_dim=model_cfg["time_emb_dim"],
        num_timesteps=model_cfg["num_timesteps"]
    ).to(device)

    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()
    print("[INFO] Model reloaded successfully.")

    scheduler = DiffusionScheduler(
        T=diffusion_cfg["T"],
        beta_start=diffusion_cfg["beta_start"],
        beta_end=diffusion_cfg["beta_end"],
        device=device
    )

    eval_folder = "/home/ubuntu-user/Desktop/code-9-8/spilt-9-27-raw/test"
    signal_length = data_cfg["signal_length"]
    #step = data_cfg.get("raw_processing", {}).get("step", 640)
    step = 640
    #cut_head = data_cfg.get("cut_head", 640)          # 读取裁剪开头参数
    cut_head = 0
    eval_start = 0                          # 评估时跳过开头的点数

    print(f"[INFO] Loading samples from: {eval_folder}")
    print(f"[INFO] cut_head = {cut_head}, eval_start = {eval_start}")

    # ========== 批量处理配置 ==========
    all_csv_files = [f for f in os.listdir(eval_folder) if f.endswith(".csv")]
    prefixes = []
    for f in all_csv_files:
        try:
            pref = int(f.split(".")[0])
            prefixes.append(pref)
        except ValueError:
            continue
    prefixes = sorted(prefixes)
    print(f"[INFO] 找到 {len(prefixes)} 个 CSV 文件: {prefixes[:10]}{'...' if len(prefixes)>10 else ''}")

    max_windows_per_file = 10000
    total_sample_limit = 1000000000

    sample_list = []
    for pref in prefixes:
        csv_path = os.path.join(eval_folder, f"{pref}.csv")
        try:
            data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
        except:
            continue
        if 'ECG' not in data.dtype.names or 'RCG' not in data.dtype.names:
            continue
        ecg_all = data['ECG']
        rcg_all = data['RCG']
        if len(ecg_all) != len(rcg_all):
            continue
        # 裁剪开头后的长度
        cropped_len = max(0, len(ecg_all) - cut_head)
        num_windows = max(0, (cropped_len - signal_length) // step + 1) if cropped_len >= signal_length else 0
        if num_windows <= 0:
            continue
        if max_windows_per_file is not None:
            indices = range(min(max_windows_per_file, num_windows))
        else:
            indices = range(num_windows)
        for idx in indices:
            sample_list.append((pref, idx))
            if total_sample_limit is not None and len(sample_list) >= total_sample_limit:
                break
        if total_sample_limit is not None and len(sample_list) >= total_sample_limit:
            break

    print(f"[INFO] 共生成 {len(sample_list)} 个待处理样本")

    # ========== 存储结果 ==========
    results = []
    ecg_list, rcg_true_list, rcg_pred_list = [], [], []

    for pref, idx in sample_list:
        try:
            rcg_np, ecg_np = load_single_sample(eval_folder, pref, idx, signal_length, step, cut_head)
        except (FileNotFoundError, IndexError, RuntimeError) as e:
            print(f"[WARN] 跳过 ({pref}, {idx})：{e}")
            continue

        rcg_cond = torch.from_numpy(rcg_np).unsqueeze(0).to(device)
        ecg_target = torch.from_numpy(ecg_np).unsqueeze(0).to(device)

        ecg_pred = scheduler.sample(model, rcg_cond, ecg_target.shape)

        rcg_arr = rcg_cond[0, 0].cpu().numpy()
        ecg_true_arr = ecg_target[0, 0].cpu().numpy()
        ecg_pred_arr = ecg_pred[0, 0].cpu().numpy()

        # ---------- 保存预测ECG ----------
        save_path = save_folder / f"{pref}_{idx}.csv"
        np.savetxt(save_path, ecg_pred_arr, delimiter=',', header='ECG_pred', comments='')
      # ========== 汇总统计 ==========
      # ========== 将原来保存 JSON 的部分替换为直接打印 ==========
    if results:
      corr_values = [r["correlation"] for r in results]
      mse_values = [r["mse"] for r in results]
      mae_values = [r["mae"] for r in results]
      freq_diffs = [r["freq_diff"] for r in results]
      abs_freq_diffs = [r["abs_freq_diff"] for r in results]

      count_le_8 = sum(1 for d in abs_freq_diffs if d <= 8.0)
      count_le_12 = sum(1 for d in abs_freq_diffs if d <= 12.0)
      total = len(results)
      prob_le_8 = count_le_8 / total
      prob_le_12 = count_le_12 / total

      summary_stats = {
        "num_samples": total,
        "correlation_mean": float(np.mean(corr_values)),
        "correlation_std": float(np.std(corr_values)),
        "correlation_min": float(np.min(corr_values)),
        "correlation_max": float(np.max(corr_values)),
        "mse_mean": float(np.mean(mse_values)),
        "mse_std": float(np.std(mse_values)),
        "mae_mean": float(np.mean(mae_values)),
        "mae_std": float(np.std(mae_values)),
        "freq_diff_mean": float(np.mean(freq_diffs)),
        "freq_diff_std": float(np.std(freq_diffs)),
        "abs_freq_diff_mean": float(np.mean(abs_freq_diffs)),
        "abs_freq_diff_std": float(np.std(abs_freq_diffs)),
        "count_abs_le_8": count_le_8,
        "prob_abs_le_8": prob_le_8,
        "count_abs_le_12": count_le_12,
        "prob_abs_le_12": prob_le_12,
      }

      print("\n========== 汇总统计 ==========")
      for k, v in summary_stats.items():
        print(f"{k}: {v:.6f}")

      # 绘图部分也相应调整标题（可选）
      num_plot = min(len(ecg_list), 4)
      if num_plot > 0:
        fig, axes = plt.subplots(num_plot, 3, figsize=(15, 4*num_plot))
        if num_plot == 1:
            axes = axes.reshape(1, -1)
        for i in range(num_plot):
            axes[i, 0].plot(ecg_list[i])
            axes[i, 0].set_title(f"Input RCG (sample {i+1})")   # 第一列改为 RCG 条件
            axes[i, 1].plot(rcg_true_list[i], label='True ECG')
            axes[i, 1].plot(rcg_pred_list[i], alpha=0.7, label='Pred ECG')
            axes[i, 1].axvline(x=eval_start, color='red', linestyle='--', label='eval start')
            axes[i, 1].set_title("True ECG vs Pred ECG")
            axes[i, 1].legend()
            axes[i, 2].plot(rcg_pred_list[i], label='Pred ECG')
            axes[i, 2].set_title(f"Pred ECG (corr={results[i]['correlation']:.3f})")
            axes[i, 2].axvline(x=eval_start, color='red', linestyle='--')
        plt.tight_layout()
        plt.show()
        print("[INFO] 已显示样本对比图")

if __name__ == "__main__":
    main()
```

```python
import os
import json
import math
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import random
def set_seed(seed=0):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(0)
# =========================================================
# 0. 路径与配置
# =========================================================
ART_DIR = Path("/home/ubuntu-user/Desktop/code-9-8/our-9-27-raw")
CKPT_PATH = ART_DIR / "last_model.pt"
TRAIN_CFG_PATH = ART_DIR / "train_config.json"
DATA_CFG_PATH = ART_DIR / "data_config.json"

# 评估时从信号的第几个点开始计算指标（跳过开头）
EVAL_START = 100   # 可修改，与训练时的 cut_head 一致
SAMPLING_RATE = 200
SAVE_DIR = ART_DIR / "single_infer_last"
SAVE_DIR.mkdir(parents=True, exist_ok=True)


def sliding_window_cut(data, window_len, step=30):
    """
    将一维数据 data 切成长度为 window_len 的片段，步长为 step。
    返回所有片段的列表。
    """
    if len(data) < window_len:
        return []
    indices = range(0, len(data) - window_len + 1, step)
    return [data[i:i+window_len] for i in indices]


# =========================================================
# 1. 与训练一致的模型定义
# =========================================================
def sinusoidal_time_embedding(timesteps, embed_dim):
    device = timesteps.device
    half_dim = embed_dim // 2
    exp_range = torch.arange(half_dim, device=device, dtype=torch.float32)
    exp_range = -math.log(10000.0) * exp_range / float(half_dim)
    sinusoid_inp = timesteps.float()[:, None] * torch.exp(exp_range)[None, :]
    emb = torch.cat([torch.sin(sinusoid_inp), torch.cos(sinusoid_inp)], dim=1)
    return emb


class SqueezeExcitation(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        reduced_channels = max(1, channels // reduction)
        self.avg_pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, reduced_channels),
            nn.SiLU(),
            nn.Linear(reduced_channels, channels),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _ = x.size()
        y = self.avg_pool(x).view(b, c)
        y = self.fc(y).view(b, c, 1)
        return x * y


class SelfAttention1D(nn.Module):
    def __init__(self, channels):
        super().__init__()

        if channels < 8:
            num_heads = 1
        elif channels % 8 == 0:
            num_heads = 8
        else:
            num_heads = max(1, channels // 2)

        if channels >= 8 and channels % 8 == 0:
            groups = 8
        elif channels >= 4 and channels % 4 == 0:
            groups = 4
        elif channels >= 2 and channels % 2 == 0:
            groups = 2
        else:
            groups = 1

        self.norm = nn.GroupNorm(num_groups=groups, num_channels=channels)
        self.attn = nn.MultiheadAttention(
            embed_dim=channels,
            num_heads=num_heads,
            batch_first=True
        )

    def forward(self, x):
        h = self.norm(x)
        h = h.permute(0, 2, 1)
        attn_out, _ = self.attn(h, h, h)
        attn_out = attn_out.permute(0, 2, 1)
        return x + attn_out


class ResidualBlock(nn.Module):
    def __init__(self, in_channels, out_channels, time_emb_dim=None):
        super().__init__()

        def make_norm(channels):
            if channels >= 8 and channels % 8 == 0:
                groups = 8
            elif channels >= 4 and channels % 4 == 0:
                groups = 4
            elif channels >= 2 and channels % 2 == 0:
                groups = 2
            else:
                groups = 1
            return nn.GroupNorm(num_groups=groups, num_channels=channels)

        self.norm1 = make_norm(in_channels)
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size=3, padding=1)

        self.time_fc = nn.Linear(time_emb_dim, out_channels) if time_emb_dim is not None else None

        self.norm2 = make_norm(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size=3, padding=1)

        self.se = SqueezeExcitation(out_channels)
        self.shortcut = nn.Conv1d(in_channels, out_channels, kernel_size=1) if in_channels != out_channels else None
        self.act = nn.SiLU()

    def forward(self, x, time_emb=None):
        h = self.norm1(x)
        h = self.act(h)
        h = self.conv1(h)

        if self.time_fc is not None and time_emb is not None:
            time_bias = self.time_fc(time_emb)
            h = h + time_bias[..., None]

        h = self.norm2(h)
        h = self.act(h)
        h = self.conv2(h)

        h = self.se(h)

        x_short = self.shortcut(x) if self.shortcut is not None else x
        return x_short + h


class ConditionalDiffusionModel(nn.Module):
    def __init__(self, in_channels=2, base_channels=64, time_emb_dim=128, num_timesteps=100):
        super().__init__()
        self.in_channels = in_channels
        self.base_channels = base_channels
        self.time_emb_dim = time_emb_dim
        self.num_timesteps = num_timesteps

        time_mlp_dim = base_channels * 8
        self.time_mlp = nn.Sequential(
            nn.Linear(time_emb_dim, time_mlp_dim),
            nn.SiLU(),
            nn.Linear(time_mlp_dim, time_mlp_dim)
        )

        self.init_conv = nn.Conv1d(in_channels, base_channels, kernel_size=3, padding=1)

        self.down_blocks = nn.ModuleList()
        self.downsamples = nn.ModuleList()

        down_channels = [
            base_channels,
            base_channels * 2,
            base_channels * 4,
            base_channels * 8,
            base_channels * 8,
            base_channels * 8
        ]

        in_ch = base_channels
        for i, out_ch in enumerate(down_channels):
            self.down_blocks.append(
                ResidualBlock(in_ch, out_ch, time_emb_dim=time_mlp_dim)
            )
            if i != len(down_channels) - 1:
                self.downsamples.append(
                    nn.Conv1d(out_ch, out_ch, kernel_size=4, stride=2, padding=1)
                )
            in_ch = out_ch

        self.attn = SelfAttention1D(down_channels[-1])

        self.up_blocks = nn.ModuleList()
        self.upsamples = nn.ModuleList()

        up_channels = [
            base_channels * 8,
            base_channels * 4,
            base_channels * 2,
            base_channels,
            base_channels
        ]

        in_ch = down_channels[-1]
        for i, out_ch in enumerate(up_channels):
            skip_ch = down_channels[-2 - i]
            res_in = in_ch + skip_ch
            self.up_blocks.append(
                ResidualBlock(res_in, out_ch, time_emb_dim=time_mlp_dim)
            )
            if i != len(up_channels) - 1:
                self.upsamples.append(
                    nn.ConvTranspose1d(out_ch, out_ch, kernel_size=4, stride=2, padding=1)
                )
            in_ch = out_ch

        self.final_conv = nn.Conv1d(up_channels[-1], 1, kernel_size=3, padding=1)

    def forward(self, x_noisy, cond, t):
        x = torch.cat([cond, x_noisy], dim=1)
        t_embed = sinusoidal_time_embedding(t, self.time_emb_dim)
        t_embed = self.time_mlp(t_embed)

        skips = []
        h = self.init_conv(x)

        for i, block in enumerate(self.down_blocks):
            h = block(h, time_emb=t_embed)
            if i != len(self.down_blocks) - 1:
                skips.append(h)
                h = self.downsamples[i](h)

        h = self.attn(h)

        for i, block in enumerate(self.up_blocks):
            skip_feat = skips[-1 - i]
            if h.size(-1) != skip_feat.size(-1):
                h = nn.functional.interpolate(h, size=skip_feat.size(-1), mode="nearest")
            h = torch.cat([h, skip_feat], dim=1)
            h = block(h, time_emb=t_embed)
            if i != len(self.up_blocks) - 1:
                h = self.upsamples[i](h)

        return self.final_conv(h)


class DiffusionScheduler:
    def __init__(self, T, beta_start, beta_end, device):
        self.T = T
        self.device = device
        self.betas = torch.linspace(beta_start, beta_end, T, device=device)
        self.alphas = 1.0 - self.betas
        self.alpha_cum = torch.cumprod(self.alphas, dim=0)

    def sample(self, model, cond, shape):
        model.eval()
        x = torch.randn(shape, device=self.device)

        with torch.no_grad():
            for t_step in reversed(range(self.T)):
                t_cur = torch.full((shape[0],), t_step, device=self.device, dtype=torch.long)
                eps_pred = model(x, cond, t_cur)

                alpha_t = self.alphas[t_step]
                alpha_bar_t = self.alpha_cum[t_step]

                mean = (1.0 / torch.sqrt(alpha_t)) * (
                    x - (1.0 - alpha_t) / torch.sqrt(1.0 - alpha_bar_t) * eps_pred
                )

                if t_step == 0:
                    x = mean
                else:
                    posterior_var = self.betas[t_step] * (1.0 - self.alpha_cum[t_step - 1]) / (1.0 - alpha_bar_t)
                    noise = torch.randn_like(x)
                    x = mean + torch.sqrt(posterior_var) * noise
        return x


# =========================================================
# 2. 读单条样本（支持裁剪开头）
# =========================================================
# =========================================================
# 2. 读单条样本（返回条件 RCG，目标 ECG）
# =========================================================
# 2. 读单条样本（返回条件 RCG，目标 ECG）
# =========================================================
def load_single_sample(eval_folder, prefix, index, window_len=640, step=30, cut_head=100):
    csv_path = os.path.join(eval_folder, f"{prefix}.csv")
    if not os.path.exists(csv_path):
        raise FileNotFoundError(f"找不到 {csv_path}")

    try:
        data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
    except Exception as e:
        raise RuntimeError(f"读取 {csv_path} 失败: {e}")

    if 'ECG' not in data.dtype.names or 'RCG' not in data.dtype.names:
        raise ValueError(f"{csv_path} 中缺少 'ECG' 或 'RCG' 列")

    ecg_all = data['ECG']
    rcg_all = data['RCG']

    if len(ecg_all) != len(rcg_all):
        raise ValueError(f"{csv_path} 中 ECG 和 RCG 长度不一致")

    if len(ecg_all) > cut_head:
        ecg_all = ecg_all[cut_head:]
        rcg_all = rcg_all[cut_head:]
    else:
        raise RuntimeError(f"{csv_path} 长度 {len(ecg_all)} <= {cut_head}，无法裁剪")

    ecg_windows = sliding_window_cut(ecg_all, window_len, step)
    rcg_windows = sliding_window_cut(rcg_all, window_len, step)

    if len(ecg_windows) == 0:
        raise RuntimeError(f"{csv_path} 裁剪后长度不足 {window_len}，无法切出窗口")

    if index < 0 or index >= len(ecg_windows):
        raise IndexError(f"TARGET_INDEX={index} 超出范围，该文件共有 {len(ecg_windows)} 个窗口")

    rcg = rcg_windows[index]
    ecg = ecg_windows[index]

    if rcg.ndim == 1:
        rcg = rcg[None, :]
    if ecg.ndim == 1:
        ecg = ecg[None, :]

    # 返回 (条件 RCG, 目标 ECG)
    return rcg.astype(np.float32), ecg.astype(np.float32)

def get_peak_freq(signal, fs):
    """
    计算信号幅度谱峰值对应的频率。
    参数：
        signal: 一维数组
        fs: 采样率 (Hz)
    返回：
        峰值频率 (单位 Hz)
    """
    n = len(signal)
    if n < 2:
        return 0.0
    fft_vals = np.fft.rfft(signal)
    magnitudes = np.abs(fft_vals)
    magnitudes[0] = 0.0          # 忽略直流分量
    peak_idx = np.argmax(magnitudes)
    freqs = np.fft.rfftfreq(n, d=1/fs)
    return freqs[peak_idx]

# =========================================================
# 3. 主流程
# =========================================================
def main():
    assert CKPT_PATH.exists(), f"找不到 checkpoint: {CKPT_PATH}"
    assert TRAIN_CFG_PATH.exists(), f"找不到配置文件: {TRAIN_CFG_PATH}"
    assert DATA_CFG_PATH.exists(), f"找不到配置文件: {DATA_CFG_PATH}"

    save_folder = Path("/home/ubuntu-user/Desktop/code-9-8/generate-9-27-IB")
    save_folder.mkdir(parents=True, exist_ok=True)
    print(f"[INFO] 生成的数据将保存到: {save_folder}")

    with open(TRAIN_CFG_PATH, "r", encoding="utf-8") as f:
        train_cfg = json.load(f)

    with open(DATA_CFG_PATH, "r", encoding="utf-8") as f:
        data_cfg = json.load(f)

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print("[INFO] Using device:", device)

    ckpt = torch.load(CKPT_PATH, map_location=device)
    model_cfg = ckpt["model_config"]
    diffusion_cfg = ckpt["diffusion_config"]

    print("[INFO] Loading checkpoint from:", CKPT_PATH)
    print("[INFO] model_config:", model_cfg)
    print("[INFO] diffusion_config:", diffusion_cfg)
    print("[INFO] checkpoint epoch:", ckpt.get("epoch", "unknown"))

    model = ConditionalDiffusionModel(
        in_channels=model_cfg["in_channels"],
        base_channels=model_cfg["base_channels"],
        time_emb_dim=model_cfg["time_emb_dim"],
        num_timesteps=model_cfg["num_timesteps"]
    ).to(device)

    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()
    print("[INFO] Model reloaded successfully.")

    scheduler = DiffusionScheduler(
        T=diffusion_cfg["T"],
        beta_start=diffusion_cfg["beta_start"],
        beta_end=diffusion_cfg["beta_end"],
        device=device
    )

    eval_folder = "/home/ubuntu-user/Desktop/code-9-8/spilt-9-27-raw/finalpartialIB15_raw"
    signal_length = data_cfg["signal_length"]
    #step = data_cfg.get("raw_processing", {}).get("step", 640)
    step = 640
    #cut_head = data_cfg.get("cut_head", 640)          # 读取裁剪开头参数
    cut_head = 0
    eval_start = 0                          # 评估时跳过开头的点数

    print(f"[INFO] Loading samples from: {eval_folder}")
    print(f"[INFO] cut_head = {cut_head}, eval_start = {eval_start}")

    # ========== 批量处理配置 ==========
    all_csv_files = [f for f in os.listdir(eval_folder) if f.endswith(".csv")]
    prefixes = []
    for f in all_csv_files:
        try:
            pref = int(f.split(".")[0])
            prefixes.append(pref)
        except ValueError:
            continue
    prefixes = sorted(prefixes)
    print(f"[INFO] 找到 {len(prefixes)} 个 CSV 文件: {prefixes[:10]}{'...' if len(prefixes)>10 else ''}")

    max_windows_per_file = 10000
    total_sample_limit = 1000000000

    sample_list = []
    for pref in prefixes:
        csv_path = os.path.join(eval_folder, f"{pref}.csv")
        try:
            data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
        except:
            continue
        if 'ECG' not in data.dtype.names or 'RCG' not in data.dtype.names:
            continue
        ecg_all = data['ECG']
        rcg_all = data['RCG']
        if len(ecg_all) != len(rcg_all):
            continue
        # 裁剪开头后的长度
        cropped_len = max(0, len(ecg_all) - cut_head)
        num_windows = max(0, (cropped_len - signal_length) // step + 1) if cropped_len >= signal_length else 0
        if num_windows <= 0:
            continue
        if max_windows_per_file is not None:
            indices = range(min(max_windows_per_file, num_windows))
        else:
            indices = range(num_windows)
        for idx in indices:
            sample_list.append((pref, idx))
            if total_sample_limit is not None and len(sample_list) >= total_sample_limit:
                break
        if total_sample_limit is not None and len(sample_list) >= total_sample_limit:
            break

    print(f"[INFO] 共生成 {len(sample_list)} 个待处理样本")

    # ========== 存储结果 ==========
    results = []
    ecg_list, rcg_true_list, rcg_pred_list = [], [], []

    for pref, idx in sample_list:
        try:
            rcg_np, ecg_np = load_single_sample(eval_folder, pref, idx, signal_length, step, cut_head)
        except (FileNotFoundError, IndexError, RuntimeError) as e:
            print(f"[WARN] 跳过 ({pref}, {idx})：{e}")
            continue

        rcg_cond = torch.from_numpy(rcg_np).unsqueeze(0).to(device)
        ecg_target = torch.from_numpy(ecg_np).unsqueeze(0).to(device)

        ecg_pred = scheduler.sample(model, rcg_cond, ecg_target.shape)

        rcg_arr = rcg_cond[0, 0].cpu().numpy()
        ecg_true_arr = ecg_target[0, 0].cpu().numpy()
        ecg_pred_arr = ecg_pred[0, 0].cpu().numpy()

        # ---------- 保存预测ECG ----------
        save_path = save_folder / f"{pref}_{idx}.csv"
        np.savetxt(save_path, ecg_pred_arr, delimiter=',', header='ECG_pred', comments='')
      # ========== 汇总统计 ==========
      # ========== 将原来保存 JSON 的部分替换为直接打印 ==========
    if results:
      corr_values = [r["correlation"] for r in results]
      mse_values = [r["mse"] for r in results]
      mae_values = [r["mae"] for r in results]
      freq_diffs = [r["freq_diff"] for r in results]
      abs_freq_diffs = [r["abs_freq_diff"] for r in results]

      count_le_8 = sum(1 for d in abs_freq_diffs if d <= 8.0)
      count_le_12 = sum(1 for d in abs_freq_diffs if d <= 12.0)
      total = len(results)
      prob_le_8 = count_le_8 / total
      prob_le_12 = count_le_12 / total

      summary_stats = {
        "num_samples": total,
        "correlation_mean": float(np.mean(corr_values)),
        "correlation_std": float(np.std(corr_values)),
        "correlation_min": float(np.min(corr_values)),
        "correlation_max": float(np.max(corr_values)),
        "mse_mean": float(np.mean(mse_values)),
        "mse_std": float(np.std(mse_values)),
        "mae_mean": float(np.mean(mae_values)),
        "mae_std": float(np.std(mae_values)),
        "freq_diff_mean": float(np.mean(freq_diffs)),
        "freq_diff_std": float(np.std(freq_diffs)),
        "abs_freq_diff_mean": float(np.mean(abs_freq_diffs)),
        "abs_freq_diff_std": float(np.std(abs_freq_diffs)),
        "count_abs_le_8": count_le_8,
        "prob_abs_le_8": prob_le_8,
        "count_abs_le_12": count_le_12,
        "prob_abs_le_12": prob_le_12,
      }

      print("\n========== 汇总统计 ==========")
      for k, v in summary_stats.items():
        print(f"{k}: {v:.6f}")

      # 绘图部分也相应调整标题（可选）
      num_plot = min(len(ecg_list), 4)
      if num_plot > 0:
        fig, axes = plt.subplots(num_plot, 3, figsize=(15, 4*num_plot))
        if num_plot == 1:
            axes = axes.reshape(1, -1)
        for i in range(num_plot):
            axes[i, 0].plot(ecg_list[i])
            axes[i, 0].set_title(f"Input RCG (sample {i+1})")   # 第一列改为 RCG 条件
            axes[i, 1].plot(rcg_true_list[i], label='True ECG')
            axes[i, 1].plot(rcg_pred_list[i], alpha=0.7, label='Pred ECG')
            axes[i, 1].axvline(x=eval_start, color='red', linestyle='--', label='eval start')
            axes[i, 1].set_title("True ECG vs Pred ECG")
            axes[i, 1].legend()
            axes[i, 2].plot(rcg_pred_list[i], label='Pred ECG')
            axes[i, 2].set_title(f"Pred ECG (corr={results[i]['correlation']:.3f})")
            axes[i, 2].axvline(x=eval_start, color='red', linestyle='--')
        plt.tight_layout()
        plt.show()
        print("[INFO] 已显示样本对比图")

if __name__ == "__main__":
    main()
```

训练RCG生成模型

```python
import os
import json
import math
import random
import time
from pathlib import Path

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader


# =========================================================
# 0. 全局配置
# =========================================================
CFG = {
    "seed": 0,
    "train_folder": "/home/ubuntu-user/Desktop/code-9-8/spilt-9-27-raw/train",
    "eval_folder": "/home/ubuntu-user/Desktop/code-9-8/spilt-9-27-raw/val",
    "artifact_dir": "/home/ubuntu-user/Desktop/code-9-8/unet-9-27",

    "signal_length": 640,
    "rcg_channels": 1,
    "ecg_channels": 1,

    "batch_size": 64,
    "num_epochs": 100,
    "learning_rate": 1e-4,
    "num_workers": 8,
    "eval_every": 1,
    "recon_every": 5,

    "T": 100,
    "beta_start": 1e-4,
    "beta_end": 0.02,

    "in_channels": 2,
    "base_channels": 64,
    "time_emb_dim": 128,

    "eval_subset_size": 8,
    "save_every": 10,

    # ---- attention 使用位置 ----
    # 可选值: "down1","down2","down3","bottleneck","up3","up2","up1"
    "attn_stages": ["bottleneck"],
}


# =========================================================
# 1. 随机种子
# =========================================================
def set_seed(seed=0):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(CFG["seed"])


# =========================================================
# 2. 目录准备
# =========================================================
ART_DIR = Path(CFG["artifact_dir"])
ART_DIR.mkdir(parents=True, exist_ok=True)


def fmt_hms(seconds: float) -> str:
    seconds = max(0, int(round(seconds)))
    h, rem = divmod(seconds, 3600)
    m, s = divmod(rem, 60)
    return f"{h:02d}:{m:02d}:{s:02d}"


def fmt_hms_ms(seconds: float) -> str:
    """格式化为 HH:MM:SS.ss，秒保留两位小数。"""
    seconds = max(0.0, float(seconds))
    h = int(seconds // 3600)
    m = int((seconds % 3600) // 60)
    s = seconds % 60
    return f"{h:02d}:{m:02d}:{s:05.2f}"


# =========================================================
# 3. 数据加载
# =========================================================
def load_data_from_folder(folder):
    prefixes = []
    for file in os.listdir(folder):
        if file.endswith(".csv"):
            try:
                prefix = int(file.split(".")[0])
                prefixes.append(prefix)
            except ValueError:
                print(f"[WARN] 文件 {file} 命名异常，跳过。")
    prefixes = sorted(list(set(prefixes)))

    ecg_list, rcg_list = [], []
    for prefix in prefixes:
        csv_path = os.path.join(folder, f"{prefix}.csv")
        if not os.path.exists(csv_path):
            print(f"[WARN] 文件 {csv_path} 不存在，跳过。")
            continue
        try:
            data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
        except Exception as e:
            print(f"[WARN] 读取 {csv_path} 失败：{e}，跳过。")
            continue

        if 'ECG' not in data.dtype.names or 'RCG' not in data.dtype.names:
            print(f"[WARN] {csv_path} 缺少 'ECG' 或 'RCG_ch1' 列，跳过。")
            continue

        ecg_data = data['ECG']
        rcg_data = data['RCG']

        if ecg_data.shape[0] != rcg_data.shape[0]:
            raise ValueError(f"[ERROR] {csv_path} 中 ECG 与 RCG 样本数不一致")

        ecg_list.append(ecg_data)
        rcg_list.append(rcg_data)

    if len(ecg_list) == 0:
        raise RuntimeError(f"[ERROR] 在文件夹 {folder} 中没有加载到有效数据")

    return ecg_list, rcg_list, prefixes


def sliding_window_cut(data, window_len, step=30):
    if len(data) < window_len:
        return []
    indices = range(0, len(data) - window_len + 1, step)
    return [data[i:i+window_len] for i in indices]


def ensure_channel_dim(x):
    if x.ndim == 2:
        x = x[:, None, :]
    return x


class SignalDataset(Dataset):
    def __init__(self, cond_tensor, target_tensor):
        self.cond = cond_tensor
        self.target = target_tensor

    def __len__(self):
        return len(self.cond)

    def __getitem__(self, idx):
        return self.cond[idx], self.target[idx]


# =========================================================
# 4. 时间步嵌入
# =========================================================
def sinusoidal_time_embedding(timesteps, embed_dim):
    device = timesteps.device
    half_dim = embed_dim // 2
    exp_range = torch.arange(half_dim, device=device, dtype=torch.float32)
    exp_range = -math.log(10000.0) * exp_range / float(half_dim)
    sinusoid_inp = timesteps.float()[:, None] * torch.exp(exp_range)[None, :]
    emb = torch.cat([torch.sin(sinusoid_inp), torch.cos(sinusoid_inp)], dim=1)
    return emb


# =========================================================
# 5. 模型模块
# =========================================================
class SelfAttention1D(nn.Module):
    def __init__(self, channels):
        super().__init__()

        if channels < 8:
            num_heads = 1
        elif channels % 8 == 0:
            num_heads = 8
        else:
            num_heads = max(1, channels // 2)

        if channels >= 8 and channels % 8 == 0:
            groups = 8
        elif channels >= 4 and channels % 4 == 0:
            groups = 4
        elif channels >= 2 and channels % 2 == 0:
            groups = 2
        else:
            groups = 1

        self.norm = nn.GroupNorm(num_groups=groups, num_channels=channels)
        self.attn = nn.MultiheadAttention(
            embed_dim=channels,
            num_heads=num_heads,
            batch_first=True
        )

    def forward(self, x):
        h = self.norm(x)
        h = h.permute(0, 2, 1)
        attn_out, _ = self.attn(h, h, h)
        attn_out = attn_out.permute(0, 2, 1)
        return attn_out


class ResidualBlock(nn.Module):
    """残差块：卷积 + 时间嵌入 + (可选)自注意力"""
    def __init__(self, in_channels, out_channels, time_emb_dim, use_attention=True):
        super().__init__()

        def make_norm(channels):
            if channels >= 8 and channels % 8 == 0:
                groups = 8
            elif channels >= 4 and channels % 4 == 0:
                groups = 4
            elif channels >= 2 and channels % 2 == 0:
                groups = 2
            else:
                groups = 1
            return nn.GroupNorm(num_groups=groups, num_channels=channels)

        self.norm1 = make_norm(in_channels)
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size=3, padding=1)

        self.time_fc = nn.Linear(time_emb_dim, out_channels) if time_emb_dim is not None else None

        self.norm2 = make_norm(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size=3, padding=1)

        self.attn = SelfAttention1D(out_channels) if use_attention else None

        self.shortcut = nn.Conv1d(in_channels, out_channels, kernel_size=1) if in_channels != out_channels else None
        self.act = nn.SiLU()

    def forward(self, x, time_emb=None):
        h = self.norm1(x)
        h = self.act(h)
        h = self.conv1(h)

        if self.time_fc is not None and time_emb is not None:
            time_bias = self.time_fc(time_emb)
            h = h + time_bias[..., None]

        h = self.norm2(h)
        h = self.act(h)
        h = self.conv2(h)

        if self.attn is not None:
            h = self.attn(h)

        x_short = self.shortcut(x) if self.shortcut is not None else x
        return x_short + h


class DownBlock(nn.Module):
    def __init__(self, in_channels, out_channels, time_emb_dim, use_attention=True):
        super().__init__()
        self.down = nn.Sequential(
            nn.Conv1d(in_channels, out_channels, kernel_size=4, stride=2, padding=1),
            nn.GroupNorm(1, out_channels),
            nn.SiLU()
        )
        self.res = ResidualBlock(out_channels, out_channels, time_emb_dim, use_attention=use_attention)

    def forward(self, x, time_emb):
        x = self.down(x)
        x = self.res(x, time_emb)
        return x


class UpBlock(nn.Module):
    def __init__(self, in_channels, skip_channels, out_channels, time_emb_dim, use_attention=True):
        super().__init__()
        self.up = nn.Sequential(
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv1d(in_channels, out_channels, kernel_size=3, padding=1),
            nn.GroupNorm(1, out_channels),
            nn.SiLU()
        )
        self.res = ResidualBlock(
            out_channels + skip_channels, out_channels, time_emb_dim,
            use_attention=use_attention
        )

    def forward(self, x, skip, time_emb):
        x = self.up(x)
        x = torch.cat([x, skip], dim=1)
        x = self.res(x, time_emb)
        return x


class ConditionalDiffusionModel(nn.Module):
    """基于 U-Net 的条件扩散模型（attention 位置可配置）"""

    ALL_STAGES = ("down1", "down2", "down3", "bottleneck", "up3", "up2", "up1")

    def __init__(self, in_channels=2, base_channels=64, time_emb_dim=128,
                 num_timesteps=100, attn_stages=("bottleneck",)):
        super().__init__()
        self.in_channels = in_channels
        self.base_channels = base_channels
        self.time_emb_dim = time_emb_dim
        self.num_timesteps = num_timesteps

        attn_stages = tuple(attn_stages)
        if "all" in attn_stages:
            attn_stages = self.ALL_STAGES
        self.attn_stages = attn_stages
        self._use = lambda name: name in attn_stages

        time_mlp_dim = base_channels * 8
        self.time_mlp = nn.Sequential(
            nn.Linear(time_emb_dim, time_mlp_dim),
            nn.SiLU(),
            nn.Linear(time_mlp_dim, time_mlp_dim)
        )

        self.init_conv = nn.Conv1d(in_channels, base_channels, kernel_size=3, padding=1)

        self.down1 = DownBlock(base_channels,       base_channels * 2, time_mlp_dim,
                               use_attention=self._use("down1"))
        self.down2 = DownBlock(base_channels * 2,   base_channels * 4, time_mlp_dim,
                               use_attention=self._use("down2"))
        self.down3 = DownBlock(base_channels * 4,   base_channels * 8, time_mlp_dim,
                               use_attention=self._use("down3"))

        self.bottleneck = nn.ModuleList([
            ResidualBlock(base_channels * 8, base_channels * 8, time_mlp_dim,
                          use_attention=self._use("bottleneck")),
            ResidualBlock(base_channels * 8, base_channels * 8, time_mlp_dim,
                          use_attention=self._use("bottleneck")),
        ])

        self.up3 = UpBlock(base_channels * 8, base_channels * 4, base_channels * 4, time_mlp_dim,
                           use_attention=self._use("up3"))
        self.up2 = UpBlock(base_channels * 4, base_channels * 2, base_channels * 2, time_mlp_dim,
                           use_attention=self._use("up2"))
        self.up1 = UpBlock(base_channels * 2, base_channels,     base_channels,     time_mlp_dim,
                           use_attention=self._use("up1"))

        self.final_conv = nn.Conv1d(base_channels, 1, kernel_size=3, padding=1)

    def forward(self, x_noisy, cond, t):
        x = torch.cat([cond, x_noisy], dim=1)

        t_embed = sinusoidal_time_embedding(t, self.time_emb_dim)
        t_embed = self.time_mlp(t_embed)

        h = self.init_conv(x)
        skip1 = h

        h = self.down1(h, t_embed); skip2 = h
        h = self.down2(h, t_embed); skip3 = h
        h = self.down3(h, t_embed)

        for block in self.bottleneck:
            h = block(h, t_embed)

        h = self.up3(h, skip3, t_embed)
        h = self.up2(h, skip2, t_embed)
        h = self.up1(h, skip1, t_embed)

        out = self.final_conv(h)
        return out


# =========================================================
# 6. 扩散过程
# =========================================================
class DiffusionScheduler:
    def __init__(self, T, beta_start, beta_end, device):
        self.T = T
        self.device = device
        self.betas = torch.linspace(beta_start, beta_end, T, device=device)
        self.alphas = 1.0 - self.betas
        self.alpha_cum = torch.cumprod(self.alphas, dim=0)

    def diffuse(self, x0, t):
        alpha_bar = self.alpha_cum[t].view(-1, 1, 1)
        noise = torch.randn_like(x0)
        x_t = torch.sqrt(alpha_bar) * x0 + torch.sqrt(1.0 - alpha_bar) * noise
        return x_t, noise

    def sample(self, model, cond, shape):
        model.eval()
        x = torch.randn(shape, device=self.device)
        with torch.inference_mode():
            for t_step in reversed(range(self.T)):
                t_cur = torch.full((shape[0],), t_step, device=self.device, dtype=torch.long)
                eps_pred = model(x, cond, t_cur)
                alpha_t = self.alphas[t_step]
                alpha_bar_t = self.alpha_cum[t_step]
                mean = (1.0 / torch.sqrt(alpha_t)) * (
                    x - (1.0 - alpha_t) / torch.sqrt(1.0 - alpha_bar_t) * eps_pred
                )
                if t_step == 0:
                    x = mean
                else:
                    posterior_var = self.betas[t_step] * (1.0 - self.alpha_cum[t_step - 1]) / (1.0 - alpha_bar_t)
                    noise = torch.randn_like(x)
                    x = mean + torch.sqrt(posterior_var) * noise
        return x


# =========================================================
# 7. 评估函数
# =========================================================
def evaluate_noise_mse(model, loader, scheduler, device):
    model.eval()
    total_loss, total_count = 0.0, 0
    with torch.inference_mode():
        for cond_batch, target_batch in loader:
            cond_batch = cond_batch.to(device, non_blocking=True)
            target_batch = target_batch.to(device, non_blocking=True)
            bsz = target_batch.size(0)
            t = torch.randint(0, scheduler.T, (bsz,), device=device)
            x_t, noise = scheduler.diffuse(target_batch, t)
            noise_pred = model(x_t, cond_batch, t)
            loss = nn.functional.mse_loss(noise_pred, noise, reduction="sum")
            total_loss += loss.item()
            total_count += noise.numel()
    return total_loss / total_count


def evaluate_pearson_loss(model, loader, scheduler, device):
    model.eval()
    total_loss, total_count = 0.0, 0
    with torch.inference_mode():
        for cond_batch, target_batch in loader:
            cond_batch = cond_batch.to(device, non_blocking=True)
            target_batch = target_batch.to(device, non_blocking=True)
            bsz = target_batch.size(0)
            t = torch.randint(0, scheduler.T, (bsz,), device=device)
            x_t, noise = scheduler.diffuse(target_batch, t)
            noise_pred = model(x_t, cond_batch, t)
            alpha_bar = scheduler.alpha_cum[t].view(-1, 1, 1)
            x0_pred = (x_t - torch.sqrt(1 - alpha_bar) * noise_pred) / (torch.sqrt(alpha_bar) + 1e-8)
            x_centered = x0_pred - x0_pred.mean(dim=-1, keepdim=True)
            y_centered = target_batch - target_batch.mean(dim=-1, keepdim=True)
            cov = (x_centered * y_centered).sum(dim=-1)
            std_x = torch.sqrt((x_centered ** 2).sum(dim=-1) + 1e-8)
            std_y = torch.sqrt((y_centered ** 2).sum(dim=-1) + 1e-8)
            corr = cov / (std_x * std_y)
            loss = 1 - corr.mean()
            total_loss += loss.item() * bsz
            total_count += bsz
    return total_loss / total_count if total_count else float('nan')


def evaluate_combined_loss(model, loader, scheduler, device):
    model.eval()
    total_loss, total_count = 0.0, 0
    with torch.inference_mode():
        for cond_batch, target_batch in loader:
            cond_batch = cond_batch.to(device, non_blocking=True)
            target_batch = target_batch.to(device, non_blocking=True)
            bsz = target_batch.size(0)
            t = torch.randint(0, scheduler.T, (bsz,), device=device)
            x_t, noise = scheduler.diffuse(target_batch, t)
            noise_pred = model(x_t, cond_batch, t)
            mse_loss = nn.functional.mse_loss(noise_pred, noise)
            alpha_bar = scheduler.alpha_cum[t].view(-1, 1, 1)
            x0_pred = (x_t - torch.sqrt(1 - alpha_bar) * noise_pred) / (torch.sqrt(alpha_bar) + 1e-8)
            x_centered = x0_pred - x0_pred.mean(dim=-1, keepdim=True)
            y_centered = target_batch - target_batch.mean(dim=-1, keepdim=True)
            cov = (x_centered * y_centered).sum(dim=-1)
            std_x = torch.sqrt((x_centered ** 2).sum(dim=-1) + 1e-8)
            std_y = torch.sqrt((y_centered ** 2).sum(dim=-1) + 1e-8)
            corr = cov / (std_x * std_y)
            corr_mean = corr.mean()
            loss = mse_loss + 0.1 - 0.1 * corr_mean
            total_loss += loss.item() * bsz
            total_count += bsz
    return total_loss / total_count if total_count else float('nan')


def evaluate_recon_subset(model, scheduler, cond_subset, target_subset, device):
    model.eval()
    cond_subset = cond_subset.to(device)
    target_subset = target_subset.to(device)
    pred = scheduler.sample(model, cond_subset, target_subset.shape)
    mse = nn.functional.mse_loss(pred, target_subset).item()
    mae = nn.functional.l1_loss(pred, target_subset).item()
    return mse, mae, pred.detach().cpu()


# =========================================================
# 8. 保存函数
# =========================================================
def save_json(obj, path):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(obj, f, ensure_ascii=False, indent=2)


def save_checkpoint(path, model, optimizer, epoch, best_eval_mse, cfg, data_meta):
    ckpt = {
        "epoch": epoch,
        "best_eval_mse": best_eval_mse,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "model_config": {
            "in_channels": cfg["in_channels"],
            "base_channels": cfg["base_channels"],
            "time_emb_dim": cfg["time_emb_dim"],
            "num_timesteps": cfg["T"],
            "attn_stages": list(cfg.get("attn_stages", [])),
        },
        "diffusion_config": {
            "T": cfg["T"],
            "beta_start": cfg["beta_start"],
            "beta_end": cfg["beta_end"]
        },
        "data_config": data_meta
    }
    torch.save(ckpt, path)


# =========================================================
# 9. 主函数
# =========================================================
def main():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print("[INFO] Using device:", device)

    if device.type == "cuda":
        torch.backends.cudnn.benchmark = True
        torch.backends.cuda.matmul.allow_tf32 = True
        torch.backends.cudnn.allow_tf32 = True
        torch.set_float32_matmul_precision("high")

    # 9.1 加载数据
    print("[INFO] Loading train data...")
    train_ecg_list, train_rcg_list, train_prefixes = load_data_from_folder(CFG["train_folder"])
    print("[INFO] Loading eval data...")
    eval_ecg_list, eval_rcg_list, eval_prefixes = load_data_from_folder(CFG["eval_folder"])

    window_len = CFG["signal_length"]
    step = 30

    train_ecg_samples, train_rcg_samples = [], []
    for ecg_arr, rcg_arr in zip(train_ecg_list, train_rcg_list):
        if len(ecg_arr) != len(rcg_arr):
            raise ValueError("同一文件中 ECG 和 RCG 长度不一致")
        ecg_windows = sliding_window_cut(ecg_arr, window_len, step)
        rcg_windows = sliding_window_cut(rcg_arr, window_len, step)
        if len(ecg_windows) != len(rcg_windows):
            raise ValueError("滑动窗口后数量不一致")
        train_ecg_samples.extend(ecg_windows)
        train_rcg_samples.extend(rcg_windows)

    eval_ecg_samples, eval_rcg_samples = [], []
    for ecg_arr, rcg_arr in zip(eval_ecg_list, eval_rcg_list):
        if len(ecg_arr) != len(rcg_arr):
            raise ValueError("同一文件中 ECG 和 RCG 长度不一致")
        ecg_windows = sliding_window_cut(ecg_arr, window_len, step)
        rcg_windows = sliding_window_cut(rcg_arr, window_len, step)
        if len(ecg_windows) != len(rcg_windows):
            raise ValueError("滑动窗口后数量不一致")
        eval_ecg_samples.extend(ecg_windows)
        eval_rcg_samples.extend(rcg_windows)

    train_ecg = ensure_channel_dim(np.array(train_ecg_samples))
    train_rcg = ensure_channel_dim(np.array(train_rcg_samples))
    eval_ecg  = ensure_channel_dim(np.array(eval_ecg_samples))
    eval_rcg  = ensure_channel_dim(np.array(eval_rcg_samples))

    assert train_ecg.shape[-1] == window_len
    assert train_rcg.shape[-1] == window_len
    assert eval_ecg.shape[-1] == window_len
    assert eval_rcg.shape[-1] == window_len

    train_ecg_tensor = torch.from_numpy(train_ecg).float()
    train_rcg_tensor = torch.from_numpy(train_rcg).float()
    eval_ecg_tensor  = torch.from_numpy(eval_ecg).float()
    eval_rcg_tensor  = torch.from_numpy(eval_rcg).float()

    train_dataset = SignalDataset(train_ecg_tensor, train_rcg_tensor)
    eval_dataset  = SignalDataset(eval_ecg_tensor, eval_rcg_tensor)

    train_loader = DataLoader(
        train_dataset, batch_size=CFG["batch_size"], shuffle=True,
        num_workers=CFG["num_workers"], pin_memory=(device.type == "cuda"),
        persistent_workers=(CFG["num_workers"] > 0),
        prefetch_factor=4 if CFG["num_workers"] > 0 else None
    )
    eval_loader = DataLoader(
        eval_dataset, batch_size=CFG["batch_size"], shuffle=False,
        num_workers=CFG["num_workers"], pin_memory=(device.type == "cuda"),
        persistent_workers=(CFG["num_workers"] > 0),
        prefetch_factor=4 if CFG["num_workers"] > 0 else None
    )

    print(f"[INFO] Train ECG shape: {train_ecg.shape}")
    print(f"[INFO] Train RCG shape: {train_rcg.shape}")
    print(f"[INFO] Eval ECG shape : {eval_ecg.shape}")
    print(f"[INFO] Eval RCG shape : {eval_rcg.shape}")
    print(f"[INFO] Train prefixes : {train_prefixes[:10]} ... total={len(train_prefixes)}")
    print(f"[INFO] Eval prefixes  : {eval_prefixes[:10]} ... total={len(eval_prefixes)}")

    # 9.2 保存配置
    data_meta = {
        "train_folder": CFG["train_folder"],
        "eval_folder": CFG["eval_folder"],
        "signal_length": CFG["signal_length"],
        "rcg_channels": CFG["rcg_channels"],
        "ecg_channels": CFG["ecg_channels"],
        "raw_processing": {
            "source_mat_range": "1.mat ~ 90.mat",
            "best_rcg_channel_rule": "choose channel with highest Pearson correlation to ECG",
            "window_length": 640,
            "step": 30,
            "note": "current eval set is reused as validation/test because of data amount",
            "input_output": "ECG -> RCG"
        }
    }
    save_json(CFG, ART_DIR / "train_config.json")
    save_json(data_meta, ART_DIR / "data_config.json")

    # 9.3 模型、优化器、调度器
    model = ConditionalDiffusionModel(
        in_channels=CFG["in_channels"],
        base_channels=CFG["base_channels"],
        time_emb_dim=CFG["time_emb_dim"],
        num_timesteps=CFG["T"],
        attn_stages=CFG["attn_stages"],
    ).to(device)

    print(f"[INFO] attention used at stages: {model.attn_stages}")
    n_params = sum(p.numel() for p in model.parameters())
    print(f"[INFO] total parameters: {n_params:,}")

    optimizer = optim.Adam(model.parameters(), lr=CFG["learning_rate"], foreach=True)
    scheduler = DiffusionScheduler(
        T=CFG["T"], beta_start=CFG["beta_start"],
        beta_end=CFG["beta_end"], device=device
    )

    use_amp = (device.type == "cuda")
    scaler = torch.amp.GradScaler("cuda", enabled=use_amp)

    resume_path = ART_DIR / "last_model.pt"
    start_epoch = 1
    best_eval_noise_mse = float("inf")

    if resume_path.exists():
        print(f"[INFO] Loading checkpoint from {resume_path}")
        ckpt = torch.load(resume_path, map_location=device, weights_only=False)
        model.load_state_dict(ckpt["model_state_dict"])
        optimizer.load_state_dict(ckpt["optimizer_state_dict"])
        start_epoch = ckpt["epoch"] + 1
        best_eval_noise_mse = ckpt["best_eval_mse"]
        print(f"[INFO] Resumed from epoch {ckpt['epoch']}, best_eval_noise_mse={best_eval_noise_mse:.6f}")
    else:
        print("[INFO] No checkpoint found, starting from scratch.")

    eval_subset_size = min(CFG["eval_subset_size"], len(eval_dataset))
    eval_subset_cond   = eval_ecg_tensor[:eval_subset_size].clone()
    eval_subset_target = eval_rcg_tensor[:eval_subset_size].clone()
    print(f"[INFO] Fixed eval subset size: {eval_subset_size}")

    # 9.4 训练循环
    best_val_loss = best_eval_noise_mse
    history = []
    total_epochs  = CFG["num_epochs"]
    epochs_to_run = total_epochs - start_epoch + 1

    run_start_time    = time.time()
    epoch_times_train = []
    epoch_times_total = []

    for epoch in range(start_epoch, total_epochs + 1):
        epoch_start = time.time()

        model.train()
        running_loss, running_count = 0.0, 0

        train_start = time.time()
        for cond_batch, target_batch in train_loader:
            cond_batch   = cond_batch.to(device, non_blocking=True)
            target_batch = target_batch.to(device, non_blocking=True)

            bsz = target_batch.size(0)
            t = torch.randint(0, CFG["T"], (bsz,), device=device)

            optimizer.zero_grad(set_to_none=True)

            with torch.amp.autocast(device_type=device.type, dtype=torch.float16, enabled=use_amp):
                x_t, noise = scheduler.diffuse(target_batch, t)
                noise_pred = model(x_t, cond_batch, t)

                mse_loss = nn.functional.mse_loss(noise_pred, noise)

                alpha_bar = scheduler.alpha_cum[t].view(-1, 1, 1)
                x0_pred = (x_t - torch.sqrt(1 - alpha_bar) * noise_pred) / (torch.sqrt(alpha_bar) + 1e-8)

                x_centered = x0_pred - x0_pred.mean(dim=-1, keepdim=True)
                y_centered = target_batch - target_batch.mean(dim=-1, keepdim=True)
                cov   = (x_centered * y_centered).sum(dim=-1)
                std_x = torch.sqrt((x_centered ** 2).sum(dim=-1) + 1e-8)
                std_y = torch.sqrt((y_centered ** 2).sum(dim=-1) + 1e-8)
                corr  = cov / (std_x * std_y)
                corr_mean = corr.mean()
                loss = mse_loss + 0.1 - 0.1 * corr_mean

            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()

            running_loss  += loss.item() * bsz
            running_count += bsz

        if device.type == "cuda":
            torch.cuda.synchronize()
        train_time = time.time() - train_start
        train_loss = running_loss / max(running_count, 1)

        eval_start = time.time()
        eval_noise_mse    = evaluate_noise_mse(model, eval_loader, scheduler, device)
        val_combined_loss = evaluate_combined_loss(model, eval_loader, scheduler, device)

        recon_mse, recon_mae = float("nan"), float("nan")
        if epoch % CFG["recon_every"] == 0 or epoch == CFG["num_epochs"]:
            recon_mse, recon_mae, _ = evaluate_recon_subset(
                model, scheduler, eval_subset_cond, eval_subset_target, device
            )

        if device.type == "cuda":
            torch.cuda.synchronize()
        eval_time = time.time() - eval_start
        epoch_time = time.time() - epoch_start
        epoch_times_train.append(train_time)
        epoch_times_total.append(epoch_time)

        elapsed          = time.time() - run_start_time
        avg_epoch_total  = elapsed / len(epoch_times_total)
        remaining_epochs = total_epochs - epoch
        eta_seconds      = avg_epoch_total * remaining_epochs

        epoch_info = {
            "epoch": epoch,
            "train_loss": float(train_loss),
            "val_combined_loss": float(val_combined_loss),
            "eval_noise_mse": float(eval_noise_mse),
            "eval_recon_mse_subset": float(recon_mse),
            "eval_recon_mae_subset": float(recon_mae),
            "time_train_s":   float(train_time),
            "time_eval_s":    float(eval_time),
            "time_epoch_s":   float(epoch_time),
            "time_elapsed_s": float(elapsed),
            "time_eta_s":     float(eta_seconds),
            "time_train_hms":   fmt_hms(train_time),
            "time_epoch_hms":   fmt_hms(epoch_time),
            "time_elapsed_hms": fmt_hms(elapsed),
            "time_eta_hms":     fmt_hms(eta_seconds),
        }
        history.append(epoch_info)

        print(
            f"[Epoch {epoch:03d}/{total_epochs}] "
            f"train_loss={train_loss:.6f} | "
            f"val_combined_loss={val_combined_loss:.6f} | "
            f"eval_noise_mse={eval_noise_mse:.6f} | "
            f"subset_recon_mse={recon_mse:.6f} | "
            f"subset_recon_mae={recon_mae:.6f} | "
            f"train={fmt_hms(train_time)} | "
            f"epoch={fmt_hms(epoch_time)} | "
            f"elapsed={fmt_hms(elapsed)} | "
            f"eta={fmt_hms(eta_seconds)}"
        )

        if val_combined_loss < best_val_loss:
            best_val_loss = val_combined_loss
            save_checkpoint(
                ART_DIR / "best_model.pt",
                model, optimizer, epoch, best_val_loss, CFG, data_meta
            )
            print(f"[INFO] Best model updated val_combined_loss={best_val_loss:.6f}")

        save_checkpoint(
            ART_DIR / "last_model.pt",
            model=model, optimizer=optimizer, epoch=epoch,
            best_eval_mse=best_val_loss, cfg=CFG, data_meta=data_meta
        )

        if epoch % CFG["save_every"] == 0:
            save_json(history, ART_DIR / "train_history.json")

    total_time = time.time() - run_start_time
    total_train_time = float(sum(epoch_times_train))
    total_epoch_time = float(sum(epoch_times_total))
    total_wall_time  = float(total_time)

    timing_summary = {
        "epochs_this_run": epochs_to_run,
        "first_epoch": start_epoch,
        "last_epoch": total_epochs,
        "total_time_s": total_wall_time,
        "total_time_hms": fmt_hms(total_wall_time),
        "total_train_time_s": total_train_time,
        "total_train_time_hms": fmt_hms(total_train_time),
        "total_epoch_time_s": total_epoch_time,
        "total_epoch_time_hms": fmt_hms(total_epoch_time),
        "avg_epoch_time_s": float(sum(epoch_times_total) / max(len(epoch_times_total), 1)),
        "avg_train_time_s": float(sum(epoch_times_train) / max(len(epoch_times_train), 1)),
        "min_epoch_time_s": float(min(epoch_times_total)) if epoch_times_total else 0.0,
        "max_epoch_time_s": float(max(epoch_times_total)) if epoch_times_total else 0.0,
        "attn_stages": list(CFG["attn_stages"]),
    }
    save_json(history, ART_DIR / "train_history.json")
    save_json(timing_summary, ART_DIR / "timing_summary.json")

    # ---- 最终输出（按你要求的格式）----
    print("[INFO] Training complete.")
    print(f"[INFO] Best validation combined loss: {best_val_loss:.6f}")
    print(f"[INFO] Total training time (sum of pure train time): {fmt_hms_ms(total_train_time)}")
    print(f"[INFO] Total epoch time (sum of train+val time)    : {fmt_hms_ms(total_epoch_time)}")
    print(f"[INFO] Total wall time (from start to end)          : {fmt_hms_ms(total_wall_time)}")
    print(f"[INFO] Artifacts saved to: {ART_DIR.resolve()}")


if __name__ == "__main__":
    main()
```

```python
import os
import json
import math
import re
from pathlib import Path
import random
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
def set_seed(seed=0):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)

# =========================================================
# 0. 路径与配置（请修改为实际路径）
# =========================================================
ART_DIR = Path("/home/ubuntu-user/Desktop/code-9-8/unet-9-27")   # 训练保存目录
CKPT_PATH = ART_DIR / "last_model.pt"                                # 或 best_model.pt
TRAIN_CFG_PATH = ART_DIR / "train_config.json"
DATA_CFG_PATH = ART_DIR / "data_config.json"

# 评估参数
EVAL_START = 0               # 计算指标时跳过的起始点数（通常为0）
SAVE_DIR = ART_DIR / "inference_eval_windowed"
SAVE_DIR.mkdir(parents=True, exist_ok=True)

# 数据路径
PRED_FOLDER = "/home/ubuntu-user/Desktop/code-9-8/generate-9-27"   # 存放 1_0.csv 等文件（只有ECG列）
REAL_FOLDER =  "/home/ubuntu-user/Desktop/code-9-8/spilt-9-27-raw/test"  # 原始长信号文件

SIGNAL_LENGTH = 640
STEP = 640                    # 窗口无重叠（与切片时一致）

# =========================================================
# 1. 模型定义（U‑Net 版本，与训练代码完全一致）
# =========================================================
def sinusoidal_time_embedding(timesteps, embed_dim):
    device = timesteps.device
    half_dim = embed_dim // 2
    exp_range = torch.arange(half_dim, device=device, dtype=torch.float32)
    exp_range = -math.log(10000.0) * exp_range / float(half_dim)
    sinusoid_inp = timesteps.float()[:, None] * torch.exp(exp_range)[None, :]
    emb = torch.cat([torch.sin(sinusoid_inp), torch.cos(sinusoid_inp)], dim=1)
    return emb


# =========================================================
# 5. 模型模块（自注意力变成可选）
# =========================================================
class SelfAttention1D(nn.Module):
    def __init__(self, channels):
        super().__init__()

        if channels < 8:
            num_heads = 1
        elif channels % 8 == 0:
            num_heads = 8
        else:
            num_heads = max(1, channels // 2)

        if channels >= 8 and channels % 8 == 0:
            groups = 8
        elif channels >= 4 and channels % 4 == 0:
            groups = 4
        elif channels >= 2 and channels % 2 == 0:
            groups = 2
        else:
            groups = 1

        self.norm = nn.GroupNorm(num_groups=groups, num_channels=channels)
        self.attn = nn.MultiheadAttention(
            embed_dim=channels,
            num_heads=num_heads,
            batch_first=True
        )

    def forward(self, x):
        h = self.norm(x)
        h = h.permute(0, 2, 1)
        attn_out, _ = self.attn(h, h, h)
        attn_out = attn_out.permute(0, 2, 1)
        return attn_out


class ResidualBlock(nn.Module):
    """残差块：卷积 + 时间嵌入 + (可选)自注意力"""
    def __init__(self, in_channels, out_channels, time_emb_dim, use_attention=True):
        super().__init__()

        def make_norm(channels):
            if channels >= 8 and channels % 8 == 0:
                groups = 8
            elif channels >= 4 and channels % 4 == 0:
                groups = 4
            elif channels >= 2 and channels % 2 == 0:
                groups = 2
            else:
                groups = 1
            return nn.GroupNorm(num_groups=groups, num_channels=channels)

        self.norm1 = make_norm(in_channels)
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size=3, padding=1)

        self.time_fc = nn.Linear(time_emb_dim, out_channels) if time_emb_dim is not None else None

        self.norm2 = make_norm(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size=3, padding=1)

        # 只在需要时创建 attention 模块（省参数、省显存）
        self.attn = SelfAttention1D(out_channels) if use_attention else None

        self.shortcut = nn.Conv1d(in_channels, out_channels, kernel_size=1) if in_channels != out_channels else None
        self.act = nn.SiLU()

    def forward(self, x, time_emb=None):
        h = self.norm1(x)
        h = self.act(h)
        h = self.conv1(h)

        if self.time_fc is not None and time_emb is not None:
            time_bias = self.time_fc(time_emb)
            h = h + time_bias[..., None]

        h = self.norm2(h)
        h = self.act(h)
        h = self.conv2(h)

        if self.attn is not None:
            h = self.attn(h)

        x_short = self.shortcut(x) if self.shortcut is not None else x
        return x_short + h


class DownBlock(nn.Module):
    def __init__(self, in_channels, out_channels, time_emb_dim, use_attention=True):
        super().__init__()
        self.down = nn.Sequential(
            nn.Conv1d(in_channels, out_channels, kernel_size=4, stride=2, padding=1),
            nn.GroupNorm(1, out_channels),
            nn.SiLU()
        )
        self.res = ResidualBlock(out_channels, out_channels, time_emb_dim, use_attention=use_attention)

    def forward(self, x, time_emb):
        x = self.down(x)
        x = self.res(x, time_emb)
        return x


class UpBlock(nn.Module):
    def __init__(self, in_channels, skip_channels, out_channels, time_emb_dim, use_attention=True):
        super().__init__()
        self.up = nn.Sequential(
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv1d(in_channels, out_channels, kernel_size=3, padding=1),
            nn.GroupNorm(1, out_channels),
            nn.SiLU()
        )
        self.res = ResidualBlock(
            out_channels + skip_channels, out_channels, time_emb_dim,
            use_attention=use_attention
        )

    def forward(self, x, skip, time_emb):
        x = self.up(x)
        x = torch.cat([x, skip], dim=1)
        x = self.res(x, time_emb)
        return x


class ConditionalDiffusionModel(nn.Module):
    """基于 U-Net 的条件扩散模型（attention 位置可配置）"""

    ALL_STAGES = ("down1", "down2", "down3", "bottleneck", "up3", "up2", "up1")

    def __init__(self, in_channels=2, base_channels=64, time_emb_dim=128,
                 num_timesteps=100, attn_stages=("bottleneck",)):
        super().__init__()
        self.in_channels = in_channels
        self.base_channels = base_channels
        self.time_emb_dim = time_emb_dim
        self.num_timesteps = num_timesteps

        # 处理 "all"
        attn_stages = tuple(attn_stages)
        if "all" in attn_stages:
            attn_stages = self.ALL_STAGES
        self.attn_stages = attn_stages
        self._use = lambda name: name in attn_stages

        time_mlp_dim = base_channels * 8
        self.time_mlp = nn.Sequential(
            nn.Linear(time_emb_dim, time_mlp_dim),
            nn.SiLU(),
            nn.Linear(time_mlp_dim, time_mlp_dim)
        )

        self.init_conv = nn.Conv1d(in_channels, base_channels, kernel_size=3, padding=1)

        self.down1 = DownBlock(base_channels,       base_channels * 2, time_mlp_dim,
                               use_attention=self._use("down1"))
        self.down2 = DownBlock(base_channels * 2,   base_channels * 4, time_mlp_dim,
                               use_attention=self._use("down2"))
        self.down3 = DownBlock(base_channels * 4,   base_channels * 8, time_mlp_dim,
                               use_attention=self._use("down3"))

        self.bottleneck = nn.ModuleList([
            ResidualBlock(base_channels * 8, base_channels * 8, time_mlp_dim,
                          use_attention=self._use("bottleneck")),
            ResidualBlock(base_channels * 8, base_channels * 8, time_mlp_dim,
                          use_attention=self._use("bottleneck")),
        ])

        self.up3 = UpBlock(base_channels * 8, base_channels * 4, base_channels * 4, time_mlp_dim,
                           use_attention=self._use("up3"))
        self.up2 = UpBlock(base_channels * 4, base_channels * 2, base_channels * 2, time_mlp_dim,
                           use_attention=self._use("up2"))
        self.up1 = UpBlock(base_channels * 2, base_channels,     base_channels,     time_mlp_dim,
                           use_attention=self._use("up1"))

        self.final_conv = nn.Conv1d(base_channels, 1, kernel_size=3, padding=1)

    def forward(self, x_noisy, cond, t):
        x = torch.cat([cond, x_noisy], dim=1)

        t_embed = sinusoidal_time_embedding(t, self.time_emb_dim)
        t_embed = self.time_mlp(t_embed)

        h = self.init_conv(x)
        skip1 = h

        h = self.down1(h, t_embed); skip2 = h
        h = self.down2(h, t_embed); skip3 = h
        h = self.down3(h, t_embed)

        for block in self.bottleneck:
            h = block(h, t_embed)

        h = self.up3(h, skip3, t_embed)
        h = self.up2(h, skip2, t_embed)
        h = self.up1(h, skip1, t_embed)

        out = self.final_conv(h)
        return out


# =========================================================
# 6. 扩散过程
# =========================================================
class DiffusionScheduler:
    def __init__(self, T, beta_start, beta_end, device):
        self.T = T
        self.device = device
        self.betas = torch.linspace(beta_start, beta_end, T, device=device)
        self.alphas = 1.0 - self.betas
        self.alpha_cum = torch.cumprod(self.alphas, dim=0)

    def diffuse(self, x0, t):
        alpha_bar = self.alpha_cum[t].view(-1, 1, 1)
        noise = torch.randn_like(x0)
        x_t = torch.sqrt(alpha_bar) * x0 + torch.sqrt(1.0 - alpha_bar) * noise
        return x_t, noise

    def sample(self, model, cond, shape):
        model.eval()
        x = torch.randn(shape, device=self.device)
        with torch.inference_mode():
            for t_step in reversed(range(self.T)):
                t_cur = torch.full((shape[0],), t_step, device=self.device, dtype=torch.long)
                eps_pred = model(x, cond, t_cur)
                alpha_t = self.alphas[t_step]
                alpha_bar_t = self.alpha_cum[t_step]
                mean = (1.0 / torch.sqrt(alpha_t)) * (
                    x - (1.0 - alpha_t) / torch.sqrt(1.0 - alpha_bar_t) * eps_pred
                )
                if t_step == 0:
                    x = mean
                else:
                    posterior_var = self.betas[t_step] * (1.0 - self.alpha_cum[t_step - 1]) / (1.0 - alpha_bar_t)
                    noise = torch.randn_like(x)
                    x = mean + torch.sqrt(posterior_var) * noise
        return x


# =========================================================
# 2. 加载窗口ECG和真实RCG窗口
# =========================================================
def load_ecg_window(pred_folder, prefix, idx):
    csv_path = os.path.join(pred_folder, f"{prefix}_{idx}.csv")
    if not os.path.exists(csv_path):
        raise FileNotFoundError(f"ECG文件不存在: {csv_path}")
    data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
    # 取第一列（列名可能是 ECG_pred 或其他）
    if data.dtype.names is not None:
        ecg = data[data.dtype.names[0]]
    else:
        ecg = np.genfromtxt(csv_path, delimiter=',', dtype=float, encoding='utf-8')
        ecg = ecg.flatten()
    # 确保长度
    if len(ecg) != SIGNAL_LENGTH:
        ecg = ecg[:SIGNAL_LENGTH] if len(ecg) > SIGNAL_LENGTH else np.pad(ecg, (0, SIGNAL_LENGTH - len(ecg)), 'constant')
    return ecg.astype(np.float32)

def load_real_rcg_window(real_folder, prefix, idx):
    csv_path = os.path.join(real_folder, f"{prefix}.csv")
    if not os.path.exists(csv_path):
        raise FileNotFoundError(f"真实数据文件不存在: {csv_path}")
    data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
    if 'RCG' not in data.dtype.names:
        raise ValueError(f"{csv_path} 中缺少 'RCG' 列")
    rcg_all = data['RCG']
    start = idx * STEP
    end = start + SIGNAL_LENGTH
    if end > len(rcg_all):
        raise IndexError(f"窗口 {idx} 超出信号长度 (总长度 {len(rcg_all)})")
    window = rcg_all[start:end]
    return window.astype(np.float32)

# =========================================================
# 3. 主程序
# =========================================================
def main():
    # 加载配置
    with open(TRAIN_CFG_PATH, 'r') as f:
        train_cfg = json.load(f)
    with open(DATA_CFG_PATH, 'r') as f:
        data_cfg = json.load(f)

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print("[INFO] Using device:", device)

    ckpt = torch.load(CKPT_PATH, map_location=device, weights_only=False)
    model_cfg = ckpt["model_config"]
    diffusion_cfg = ckpt["diffusion_config"]

    model = ConditionalDiffusionModel(
        in_channels=model_cfg["in_channels"],
        base_channels=model_cfg["base_channels"],
        time_emb_dim=model_cfg["time_emb_dim"],
        num_timesteps=model_cfg["num_timesteps"]
    ).to(device)
    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()
    print("[INFO] Model loaded.")

    scheduler = DiffusionScheduler(
        T=diffusion_cfg["T"],
        beta_start=diffusion_cfg["beta_start"],
        beta_end=diffusion_cfg["beta_end"],
        device=device
    )

    # 收集所有窗口文件
    ecg_files = [f for f in os.listdir(PRED_FOLDER) if f.endswith('.csv')]
    pattern = re.compile(r'^(\d+)_(\d+)\.csv$')
    sample_list = []
    for f in ecg_files:
        m = pattern.match(f)
        if m:
            sample_list.append((int(m.group(1)), int(m.group(2))))
        else:
            print(f"[WARN] 文件名 {f} 不符合预期，跳过")
    sample_list.sort(key=lambda x: (x[0], x[1]))
    print(f"[INFO] 找到 {len(sample_list)} 个ECG窗口文件")

    # 可选限制
    # sample_list = sample_list[:100]

    results = []
    ecg_list, real_list, pred_list = [], [], []

    for prefix, idx in sample_list:
        try:
            ecg_np = load_ecg_window(PRED_FOLDER, prefix, idx)
            real_np = load_real_rcg_window(REAL_FOLDER, prefix, idx)
        except Exception as e:
            print(f"[WARN] 跳过 ({prefix}, {idx})：{e}")
            continue

        # 准备模型输入
        ecg_tensor = torch.from_numpy(ecg_np).float().unsqueeze(0).unsqueeze(0).to(device)  # [1,1,L]
        real_tensor = torch.from_numpy(real_np).float().unsqueeze(0).unsqueeze(0).to(device)  # [1,1,L]

        # 生成RCG
        pred_tensor = scheduler.sample(model, ecg_tensor, real_tensor.shape)

        pred_np = pred_tensor[0, 0].cpu().numpy()
        real_np = real_tensor[0, 0].cpu().numpy()

        # 计算指标（可跳过开头）
        if EVAL_START < SIGNAL_LENGTH:
            p = pred_np[EVAL_START:]
            r = real_np[EVAL_START:]
        else:
            p, r = pred_np, real_np

        mse = np.mean((p - r) ** 2)
        mae = np.mean(np.abs(p - r))
        corr = np.corrcoef(p, r)[0, 1] if len(p) > 1 else np.nan

        results.append({
            "prefix": prefix,
            "index": idx,
            "mse": float(mse),
            "mae": float(mae),
            "correlation": float(corr) if not np.isnan(corr) else None
        })

        if len(results) <= 20:
            ecg_list.append(ecg_np)
            real_list.append(real_np)
            pred_list.append(pred_np)

        print(f"[INFO] ({prefix}, {idx}) MSE={mse:.6f}, MAE={mae:.6f}, Corr={corr:.6f}")

    # 汇总统计
    if results:
        corr_vals = [r["correlation"] for r in results if r["correlation"] is not None]
        mse_vals = [r["mse"] for r in results]
        mae_vals = [r["mae"] for r in results]

        summary = {
            "num_samples": len(results),
            "correlation_mean": np.mean(corr_vals) if corr_vals else None,
            "correlation_std": np.std(corr_vals) if corr_vals else None,
            "correlation_min": np.min(corr_vals) if corr_vals else None,
            "correlation_max": np.max(corr_vals) if corr_vals else None,
            "mse_mean": np.mean(mse_vals),
            "mse_std": np.std(mse_vals),
            "mae_mean": np.mean(mae_vals),
            "mae_std": np.std(mae_vals),
        }
        print("\n========== 汇总统计 ==========")
        for k, v in summary.items():
            print(f"{k}: {v}")

        full_results = {
            "samples": results,
            "summary": summary,
            "config": {
                "checkpoint": str(CKPT_PATH),
                "ecg_folder": PRED_FOLDER,
                "real_folder": REAL_FOLDER,
                "signal_length": SIGNAL_LENGTH,
                "step": STEP,
                "eval_start": EVAL_START
            }
        }
        with open(SAVE_DIR / "inference_eval_results.json", "w") as f:
            json.dump(full_results, f, indent=2)

        # 可视化前几个样本
        num_plot = min(len(ecg_list), 4)
        if num_plot > 0:
            fig, axes = plt.subplots(num_plot, 3, figsize=(15, 4*num_plot))
            if num_plot == 1:
                axes = axes.reshape(1, -1)
            for i in range(num_plot):
                axes[i, 0].plot(ecg_list[i])
                axes[i, 0].set_title(f"Input ECG (sample {i+1})")
                axes[i, 1].plot(real_list[i], label='True')
                axes[i, 1].plot(pred_list[i], alpha=0.7, label='Pred')
                axes[i, 1].axvline(x=EVAL_START, color='red', linestyle='--')
                axes[i, 1].set_title("True vs Pred RCG")
                axes[i, 1].legend()
                axes[i, 2].plot(pred_list[i])
                axes[i, 2].set_title(f"Pred RCG (corr={results[i]['correlation']:.3f})")
                axes[i, 2].axvline(x=EVAL_START, color='red', linestyle='--')
            plt.tight_layout()
            plt.show()
            print("[INFO] 已显示样本对比图")
    else:
        print("[ERROR] 没有成功处理任何样本")

    print("[INFO] 推理评估完成，结果保存在:", SAVE_DIR)

if __name__ == "__main__":
    main()
```

```python
import os
import json
import math
import re
from pathlib import Path
import random
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
def set_seed(seed=0):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)

# =========================================================
# 0. 路径与配置（请修改为实际路径）
# =========================================================
ART_DIR = Path("/home/ubuntu-user/Desktop/code-9-8/unet-9-27")   # 训练保存目录
CKPT_PATH = ART_DIR / "last_model.pt"                                # 或 best_model.pt
TRAIN_CFG_PATH = ART_DIR / "train_config.json"
DATA_CFG_PATH = ART_DIR / "data_config.json"

# 评估参数
EVAL_START = 0               # 计算指标时跳过的起始点数（通常为0）
SAVE_DIR = ART_DIR / "inference_eval_windowed"
SAVE_DIR.mkdir(parents=True, exist_ok=True)

# 数据路径
PRED_FOLDER = "/home/ubuntu-user/Desktop/code-9-8/generate-9-27-IB"   # 存放 1_0.csv 等文件（只有ECG列）
REAL_FOLDER = "/home/ubuntu-user/Desktop/code-9-8/spilt-9-27-raw/finalpartialIB15_raw"   # 原始长信号文件

SIGNAL_LENGTH = 640
STEP = 640                    # 窗口无重叠（与切片时一致）

# =========================================================
# 1. 模型定义（U‑Net 版本，与训练代码完全一致）
# =========================================================
def sinusoidal_time_embedding(timesteps, embed_dim):
    device = timesteps.device
    half_dim = embed_dim // 2
    exp_range = torch.arange(half_dim, device=device, dtype=torch.float32)
    exp_range = -math.log(10000.0) * exp_range / float(half_dim)
    sinusoid_inp = timesteps.float()[:, None] * torch.exp(exp_range)[None, :]
    emb = torch.cat([torch.sin(sinusoid_inp), torch.cos(sinusoid_inp)], dim=1)
    return emb


# =========================================================
# 5. 模型模块（自注意力变成可选）
# =========================================================
class SelfAttention1D(nn.Module):
    def __init__(self, channels):
        super().__init__()

        if channels < 8:
            num_heads = 1
        elif channels % 8 == 0:
            num_heads = 8
        else:
            num_heads = max(1, channels // 2)

        if channels >= 8 and channels % 8 == 0:
            groups = 8
        elif channels >= 4 and channels % 4 == 0:
            groups = 4
        elif channels >= 2 and channels % 2 == 0:
            groups = 2
        else:
            groups = 1

        self.norm = nn.GroupNorm(num_groups=groups, num_channels=channels)
        self.attn = nn.MultiheadAttention(
            embed_dim=channels,
            num_heads=num_heads,
            batch_first=True
        )

    def forward(self, x):
        h = self.norm(x)
        h = h.permute(0, 2, 1)
        attn_out, _ = self.attn(h, h, h)
        attn_out = attn_out.permute(0, 2, 1)
        return attn_out


class ResidualBlock(nn.Module):
    """残差块：卷积 + 时间嵌入 + (可选)自注意力"""
    def __init__(self, in_channels, out_channels, time_emb_dim, use_attention=True):
        super().__init__()

        def make_norm(channels):
            if channels >= 8 and channels % 8 == 0:
                groups = 8
            elif channels >= 4 and channels % 4 == 0:
                groups = 4
            elif channels >= 2 and channels % 2 == 0:
                groups = 2
            else:
                groups = 1
            return nn.GroupNorm(num_groups=groups, num_channels=channels)

        self.norm1 = make_norm(in_channels)
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size=3, padding=1)

        self.time_fc = nn.Linear(time_emb_dim, out_channels) if time_emb_dim is not None else None

        self.norm2 = make_norm(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size=3, padding=1)

        # 只在需要时创建 attention 模块（省参数、省显存）
        self.attn = SelfAttention1D(out_channels) if use_attention else None

        self.shortcut = nn.Conv1d(in_channels, out_channels, kernel_size=1) if in_channels != out_channels else None
        self.act = nn.SiLU()

    def forward(self, x, time_emb=None):
        h = self.norm1(x)
        h = self.act(h)
        h = self.conv1(h)

        if self.time_fc is not None and time_emb is not None:
            time_bias = self.time_fc(time_emb)
            h = h + time_bias[..., None]

        h = self.norm2(h)
        h = self.act(h)
        h = self.conv2(h)

        if self.attn is not None:
            h = self.attn(h)

        x_short = self.shortcut(x) if self.shortcut is not None else x
        return x_short + h


class DownBlock(nn.Module):
    def __init__(self, in_channels, out_channels, time_emb_dim, use_attention=True):
        super().__init__()
        self.down = nn.Sequential(
            nn.Conv1d(in_channels, out_channels, kernel_size=4, stride=2, padding=1),
            nn.GroupNorm(1, out_channels),
            nn.SiLU()
        )
        self.res = ResidualBlock(out_channels, out_channels, time_emb_dim, use_attention=use_attention)

    def forward(self, x, time_emb):
        x = self.down(x)
        x = self.res(x, time_emb)
        return x


class UpBlock(nn.Module):
    def __init__(self, in_channels, skip_channels, out_channels, time_emb_dim, use_attention=True):
        super().__init__()
        self.up = nn.Sequential(
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv1d(in_channels, out_channels, kernel_size=3, padding=1),
            nn.GroupNorm(1, out_channels),
            nn.SiLU()
        )
        self.res = ResidualBlock(
            out_channels + skip_channels, out_channels, time_emb_dim,
            use_attention=use_attention
        )

    def forward(self, x, skip, time_emb):
        x = self.up(x)
        x = torch.cat([x, skip], dim=1)
        x = self.res(x, time_emb)
        return x


class ConditionalDiffusionModel(nn.Module):
    """基于 U-Net 的条件扩散模型（attention 位置可配置）"""

    ALL_STAGES = ("down1", "down2", "down3", "bottleneck", "up3", "up2", "up1")

    def __init__(self, in_channels=2, base_channels=64, time_emb_dim=128,
                 num_timesteps=100, attn_stages=("bottleneck",)):
        super().__init__()
        self.in_channels = in_channels
        self.base_channels = base_channels
        self.time_emb_dim = time_emb_dim
        self.num_timesteps = num_timesteps

        # 处理 "all"
        attn_stages = tuple(attn_stages)
        if "all" in attn_stages:
            attn_stages = self.ALL_STAGES
        self.attn_stages = attn_stages
        self._use = lambda name: name in attn_stages

        time_mlp_dim = base_channels * 8
        self.time_mlp = nn.Sequential(
            nn.Linear(time_emb_dim, time_mlp_dim),
            nn.SiLU(),
            nn.Linear(time_mlp_dim, time_mlp_dim)
        )

        self.init_conv = nn.Conv1d(in_channels, base_channels, kernel_size=3, padding=1)

        self.down1 = DownBlock(base_channels,       base_channels * 2, time_mlp_dim,
                               use_attention=self._use("down1"))
        self.down2 = DownBlock(base_channels * 2,   base_channels * 4, time_mlp_dim,
                               use_attention=self._use("down2"))
        self.down3 = DownBlock(base_channels * 4,   base_channels * 8, time_mlp_dim,
                               use_attention=self._use("down3"))

        self.bottleneck = nn.ModuleList([
            ResidualBlock(base_channels * 8, base_channels * 8, time_mlp_dim,
                          use_attention=self._use("bottleneck")),
            ResidualBlock(base_channels * 8, base_channels * 8, time_mlp_dim,
                          use_attention=self._use("bottleneck")),
        ])

        self.up3 = UpBlock(base_channels * 8, base_channels * 4, base_channels * 4, time_mlp_dim,
                           use_attention=self._use("up3"))
        self.up2 = UpBlock(base_channels * 4, base_channels * 2, base_channels * 2, time_mlp_dim,
                           use_attention=self._use("up2"))
        self.up1 = UpBlock(base_channels * 2, base_channels,     base_channels,     time_mlp_dim,
                           use_attention=self._use("up1"))

        self.final_conv = nn.Conv1d(base_channels, 1, kernel_size=3, padding=1)

    def forward(self, x_noisy, cond, t):
        x = torch.cat([cond, x_noisy], dim=1)

        t_embed = sinusoidal_time_embedding(t, self.time_emb_dim)
        t_embed = self.time_mlp(t_embed)

        h = self.init_conv(x)
        skip1 = h

        h = self.down1(h, t_embed); skip2 = h
        h = self.down2(h, t_embed); skip3 = h
        h = self.down3(h, t_embed)

        for block in self.bottleneck:
            h = block(h, t_embed)

        h = self.up3(h, skip3, t_embed)
        h = self.up2(h, skip2, t_embed)
        h = self.up1(h, skip1, t_embed)

        out = self.final_conv(h)
        return out


# =========================================================
# 6. 扩散过程
# =========================================================
class DiffusionScheduler:
    def __init__(self, T, beta_start, beta_end, device):
        self.T = T
        self.device = device
        self.betas = torch.linspace(beta_start, beta_end, T, device=device)
        self.alphas = 1.0 - self.betas
        self.alpha_cum = torch.cumprod(self.alphas, dim=0)

    def diffuse(self, x0, t):
        alpha_bar = self.alpha_cum[t].view(-1, 1, 1)
        noise = torch.randn_like(x0)
        x_t = torch.sqrt(alpha_bar) * x0 + torch.sqrt(1.0 - alpha_bar) * noise
        return x_t, noise

    def sample(self, model, cond, shape):
        model.eval()
        x = torch.randn(shape, device=self.device)
        with torch.inference_mode():
            for t_step in reversed(range(self.T)):
                t_cur = torch.full((shape[0],), t_step, device=self.device, dtype=torch.long)
                eps_pred = model(x, cond, t_cur)
                alpha_t = self.alphas[t_step]
                alpha_bar_t = self.alpha_cum[t_step]
                mean = (1.0 / torch.sqrt(alpha_t)) * (
                    x - (1.0 - alpha_t) / torch.sqrt(1.0 - alpha_bar_t) * eps_pred
                )
                if t_step == 0:
                    x = mean
                else:
                    posterior_var = self.betas[t_step] * (1.0 - self.alpha_cum[t_step - 1]) / (1.0 - alpha_bar_t)
                    noise = torch.randn_like(x)
                    x = mean + torch.sqrt(posterior_var) * noise
        return x


# =========================================================
# 2. 加载窗口ECG和真实RCG窗口
# =========================================================
def load_ecg_window(pred_folder, prefix, idx):
    csv_path = os.path.join(pred_folder, f"{prefix}_{idx}.csv")
    if not os.path.exists(csv_path):
        raise FileNotFoundError(f"ECG文件不存在: {csv_path}")
    data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
    # 取第一列（列名可能是 ECG_pred 或其他）
    if data.dtype.names is not None:
        ecg = data[data.dtype.names[0]]
    else:
        ecg = np.genfromtxt(csv_path, delimiter=',', dtype=float, encoding='utf-8')
        ecg = ecg.flatten()
    # 确保长度
    if len(ecg) != SIGNAL_LENGTH:
        ecg = ecg[:SIGNAL_LENGTH] if len(ecg) > SIGNAL_LENGTH else np.pad(ecg, (0, SIGNAL_LENGTH - len(ecg)), 'constant')
    return ecg.astype(np.float32)

def load_real_rcg_window(real_folder, prefix, idx):
    csv_path = os.path.join(real_folder, f"{prefix}.csv")
    if not os.path.exists(csv_path):
        raise FileNotFoundError(f"真实数据文件不存在: {csv_path}")
    data = np.genfromtxt(csv_path, delimiter=',', names=True, dtype=float, encoding='utf-8')
    if 'RCG' not in data.dtype.names:
        raise ValueError(f"{csv_path} 中缺少 'RCG' 列")
    rcg_all = data['RCG']
    start = idx * STEP
    end = start + SIGNAL_LENGTH
    if end > len(rcg_all):
        raise IndexError(f"窗口 {idx} 超出信号长度 (总长度 {len(rcg_all)})")
    window = rcg_all[start:end]
    return window.astype(np.float32)

# =========================================================
# 3. 主程序
# =========================================================
def main():
    # 加载配置
    with open(TRAIN_CFG_PATH, 'r') as f:
        train_cfg = json.load(f)
    with open(DATA_CFG_PATH, 'r') as f:
        data_cfg = json.load(f)

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print("[INFO] Using device:", device)

    ckpt = torch.load(CKPT_PATH, map_location=device, weights_only=False)
    model_cfg = ckpt["model_config"]
    diffusion_cfg = ckpt["diffusion_config"]

    model = ConditionalDiffusionModel(
        in_channels=model_cfg["in_channels"],
        base_channels=model_cfg["base_channels"],
        time_emb_dim=model_cfg["time_emb_dim"],
        num_timesteps=model_cfg["num_timesteps"]
    ).to(device)
    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()
    print("[INFO] Model loaded.")

    scheduler = DiffusionScheduler(
        T=diffusion_cfg["T"],
        beta_start=diffusion_cfg["beta_start"],
        beta_end=diffusion_cfg["beta_end"],
        device=device
    )

    # 收集所有窗口文件
    ecg_files = [f for f in os.listdir(PRED_FOLDER) if f.endswith('.csv')]
    pattern = re.compile(r'^(\d+)_(\d+)\.csv$')
    sample_list = []
    for f in ecg_files:
        m = pattern.match(f)
        if m:
            sample_list.append((int(m.group(1)), int(m.group(2))))
        else:
            print(f"[WARN] 文件名 {f} 不符合预期，跳过")
    sample_list.sort(key=lambda x: (x[0], x[1]))
    print(f"[INFO] 找到 {len(sample_list)} 个ECG窗口文件")

    # 可选限制
    # sample_list = sample_list[:100]

    results = []
    ecg_list, real_list, pred_list = [], [], []

    for prefix, idx in sample_list:
        try:
            ecg_np = load_ecg_window(PRED_FOLDER, prefix, idx)
            real_np = load_real_rcg_window(REAL_FOLDER, prefix, idx)
        except Exception as e:
            print(f"[WARN] 跳过 ({prefix}, {idx})：{e}")
            continue

        # 准备模型输入
        ecg_tensor = torch.from_numpy(ecg_np).float().unsqueeze(0).unsqueeze(0).to(device)  # [1,1,L]
        real_tensor = torch.from_numpy(real_np).float().unsqueeze(0).unsqueeze(0).to(device)  # [1,1,L]

        # 生成RCG
        pred_tensor = scheduler.sample(model, ecg_tensor, real_tensor.shape)

        pred_np = pred_tensor[0, 0].cpu().numpy()
        real_np = real_tensor[0, 0].cpu().numpy()

        # 计算指标（可跳过开头）
        if EVAL_START < SIGNAL_LENGTH:
            p = pred_np[EVAL_START:]
            r = real_np[EVAL_START:]
        else:
            p, r = pred_np, real_np

        mse = np.mean((p - r) ** 2)
        mae = np.mean(np.abs(p - r))
        corr = np.corrcoef(p, r)[0, 1] if len(p) > 1 else np.nan

        results.append({
            "prefix": prefix,
            "index": idx,
            "mse": float(mse),
            "mae": float(mae),
            "correlation": float(corr) if not np.isnan(corr) else None
        })

        if len(results) <= 20:
            ecg_list.append(ecg_np)
            real_list.append(real_np)
            pred_list.append(pred_np)

        print(f"[INFO] ({prefix}, {idx}) MSE={mse:.6f}, MAE={mae:.6f}, Corr={corr:.6f}")

    # 汇总统计
    if results:
        corr_vals = [r["correlation"] for r in results if r["correlation"] is not None]
        mse_vals = [r["mse"] for r in results]
        mae_vals = [r["mae"] for r in results]

        summary = {
            "num_samples": len(results),
            "correlation_mean": np.mean(corr_vals) if corr_vals else None,
            "correlation_std": np.std(corr_vals) if corr_vals else None,
            "correlation_min": np.min(corr_vals) if corr_vals else None,
            "correlation_max": np.max(corr_vals) if corr_vals else None,
            "mse_mean": np.mean(mse_vals),
            "mse_std": np.std(mse_vals),
            "mae_mean": np.mean(mae_vals),
            "mae_std": np.std(mae_vals),
        }
        print("\n========== 汇总统计 ==========")
        for k, v in summary.items():
            print(f"{k}: {v}")

        full_results = {
            "samples": results,
            "summary": summary,
            "config": {
                "checkpoint": str(CKPT_PATH),
                "ecg_folder": PRED_FOLDER,
                "real_folder": REAL_FOLDER,
                "signal_length": SIGNAL_LENGTH,
                "step": STEP,
                "eval_start": EVAL_START
            }
        }
        with open(SAVE_DIR / "inference_eval_results.json", "w") as f:
            json.dump(full_results, f, indent=2)

        # 可视化前几个样本
        num_plot = min(len(ecg_list), 4)
        if num_plot > 0:
            fig, axes = plt.subplots(num_plot, 3, figsize=(15, 4*num_plot))
            if num_plot == 1:
                axes = axes.reshape(1, -1)
            for i in range(num_plot):
                axes[i, 0].plot(ecg_list[i])
                axes[i, 0].set_title(f"Input ECG (sample {i+1})")
                axes[i, 1].plot(real_list[i], label='True')
                axes[i, 1].plot(pred_list[i], alpha=0.7, label='Pred')
                axes[i, 1].axvline(x=EVAL_START, color='red', linestyle='--')
                axes[i, 1].set_title("True vs Pred RCG")
                axes[i, 1].legend()
                axes[i, 2].plot(pred_list[i])
                axes[i, 2].set_title(f"Pred RCG (corr={results[i]['correlation']:.3f})")
                axes[i, 2].axvline(x=EVAL_START, color='red', linestyle='--')
            plt.tight_layout()
            plt.show()
            print("[INFO] 已显示样本对比图")
    else:
        print("[ERROR] 没有成功处理任何样本")

    print("[INFO] 推理评估完成，结果保存在:", SAVE_DIR)

if __name__ == "__main__":
    main()
```
